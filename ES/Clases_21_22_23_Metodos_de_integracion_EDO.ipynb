## 1) Definiciones básicas

### Definición de ecuación diferencial
Una ecuación que tenga derivadas de cualquier orden en sus términos es una ecuación diferencial, por ejemplo considere las siguientes ecuaciones y su solución:
$\newcommand{\sen}{\,\hbox{sen}\,}$

$$
\begin{array}{l}
1)\quad \dfrac{dy}{dt}= -ky &\iff& y(t)=Ae^{-kt}\\
2)\quad \dfrac{d^2y}{dt^2}+ \omega_0^2y= 0  &\iff&  y(t)=A\cos(\omega_0 t) + B\sen(\omega_0 t),\\
3)\quad \dfrac{d^2y}{dt^2}+ \omega_0 \sen y= A\cos(\alpha t-\pi/3)&\iff& \text{No hay solución analítica},\\
4)\quad \dfrac{d^3y}{dt^3}+t^2y\dfrac{d^2y}{dt^2} - 3y(y+t)=0&\iff& \text{No hay solución analítica.}
\end{array}
$$

Como vemos no todas tienen solución analítica, pero existen varios métodos numéricos para resolver estas ecuaciones.

### Orden y clasificación de las ecuaciones diferenciales

El orden es determinado por la derivada más alta en la ecuación, así la ecuación 1) es de orden uno, la 2) y 3) de orden dos, y la última de orden tres. Por otro lado,
La clasificación se da de acuerdo a si la ecuación es lineal o no lineal, una EDO es lineal si la variable a integrar se puede separar del operador $\mathcal{L}$ formado por las derivadas, osea un ecuación diferencial es lineal si,

$$\mathcal{L}y(t)=f(t),$$ 

por ejemplo para la primera EDO,

$$ \left(\frac{d}{dt} + k\right)y(t)=\mathcal{L} y(t)=0.$$

Así, las ecuaciones 1) y 2) son lineales pero la 3) y 4) no.

### Sistemas de EDO de primer orden

Un sistema de $n$ ecuaciones diferenciales es lineal (es decir, no hay términos como $y_i^2, 1/y_i$ o $y_iy_j$, por ejemplo) si se puede escribir de la forma,

$$\dfrac{dy_1}{dt}=a_{11}y_1 + a_{12}y_2 + \cdots a_{1n}y_n,$$
$$\dfrac{dy_2}{dt}=a_{21}y_1 + a_{22}y_2 + \cdots a_{2n}y_n,$$
$$\vdots$$
$$\dfrac{dy_n}{dt}=a_{n1}y_1 + a_{n2}y_2 + \cdots a_{nn}y_n.$$

Si se define $\mathbf{y}(t)= (y_1(t),y_2(t),..., y_n(t))$, la forma matricial y su solución es, 

$$\quad \frac{d\mathbf{y}}{dt}= A\mathbf{y}\iff \mathbf{y}=\exp(At)\mathbf{y}_0$$

donde $\mathbf{y}_0=\mathbf{y}(t_0)$ es la condición inicial y $A$ es una matriz $n\times n$, la solución se hace por los métodos matriciales de diagolalización. Si $A(t)$ varía con el tiempo entonces la solución es, 

$$\mathbf{y}(t)=\exp\left(\int_{t_0}^tA(t')dt'\right)\mathbf{y}_0.$$

En el caso de sistemas no lineales las ecuaciones acopladas se escriben de la forma,

$$\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}),$$

note que, $\mathbf{f}(t,\mathbf{y})\neq A\mathbf{y}$, estos sistemas de EDO no tienen solución analítica (solo algunos casos) y se deben resolver de manera numérica, sistemas como el de Eduard Lorenz o el oscilador armónico para ángulos grandes son ejemplos de sistemas no lineales, en este documento se mostrarán los diferentes métodos numéricos para la solución de sistemas EDO:  Método de Euler, Métodos de verlet y leapfrog, Runge-Kuta (rk2, rk4 y rk45). Antes de ver los sistemas de EDO, primero veamos el caso de una sola variable en la siguiente sección.

## 2) Método de Euler
Considere la ecuación diferencial,

$$\frac{dy}{dt}= f(t, y),  \,\,\,   a \le t \le b, \,\,\, y(a) = \alpha.$$

Por expansión de taylor hasta el orden dos de $y(t)$ y usando la ecuación anterior para remplazar la derivada $y'(t_n)$, por $f (t_n, y(t_n))$, tenemos, 

$$y(t_{n+1}) = y(t_n) + hf (t_n, y(t_n)) + \frac{h^2}{2} y''(\xi_n),$$

por lo tanto el método de euler se define como,

$$y_{n+1} \approx y_n + hf(t_n,y_n).$$

El "*error local*" o "*error de truncación local*" mide a cada paso la cantidad por la cual la solución aproximada falla en satisfacer la ecuación diferencial y acorde a la ecuación diferencial, es definido por,

$$\boxed{\tau(h) \overset{\text{def}}{=} \frac{y_{n+1} - y_n}{h} - f(t_n,y_n)=\frac{h}{2}y''(\xi_n).}$$

por lo tanto este método es de primer orden $O(h)$.

#### Ejemplo: Péndulo 
Considere un péndulo formado por una masa $m$ y atado a una cuerda de longitud $l$, la cuerda forma un ángulo $\theta$ con la vertical, si se descompone la fuerza en las componentes radial y tangencial a la trayectoria y usamos la ecuaciones de Newton, se tiene que

$$
\begin{align}
T-mg\cos\theta =&\,\frac{v^2}{l}\\
-mg \sen\theta =& \,m\frac{d^2s}{dt^2} 
\end{align}
$$

donde $s$ es la longitud de arco recorrido, si tenemos en cuenta que $s=l\theta$ esto nos da la ecuación diferencial no lineal,

$$
 \frac{d^2\theta}{dt^2} + \omega_0^2\sen\theta = 0,
$$

donde $\omega_0 = \sqrt{\frac{g}{l}}$. Esta ecuación tiene solución analítica si se considera $\theta$ pequeño en ese caso $\sen\theta\approx \theta$ que da la ecuación del oscilador armónico (por ejemplo, note que el error relativo para $\theta=15^\circ$  grados es de ~$1.15\%$ y para para $\theta=30^\circ$ grados es de ~$5\%$, así un valor de $\theta<15^\circ$ es pequeño en el sentido de que da un error menor del uno porciento; verifique esto), sin embargo para ángulos grandes no hay solución analítica. 
Como la ecuación es de segundo orden y Euler es para ecuaciones de primer orden, primero esta se escribe como dos ecuaciones lineales de primer orden acopladas,

$$
\begin{align}
 \frac{d\theta}{dt} =&\, \omega,\\
 \frac{d\omega}{dt} =&\, - \omega_0^2\sen\theta,
\end{align}
$$

y al implementar Euler en cada ecuación,

$$
\begin{align}
 \theta_{i+1} =&\,\theta_i + h\omega_i,\\
 \omega_{i+1} =&\,\omega_i - h\omega_0^2\sen\theta_i,
\end{align}
$$

pero para una mejor convergencia se suele usar el nuevo $\theta_{i+1}$ en la segunda ecuación, ósea, 

$$\omega_{i+1} =\,\omega_i - h\omega_0^2\sen\theta_{i+1}.$$ 

Ahora veamos la solución numérica implementada con el método de Euler:

#### Ejemplo: Sistema planetario
El movimiento de los planetas se describe por la ley de gravitación universal de Newton,

$$F=-G\frac{mM}{r^2},$$

donde $G$ es la constante de gravitación universal. El movimiento de un planeta de masa $m$ con coordenadas $\mathbf{r}=(x,y)$, alrededor del sol (masa $M$) se puede describir por por la ecuación,

$$
\mathbf{F} = m\mathbf{a}=m\frac{d^2\mathbf{r}}{dt^2},
$$

cuyas componentes son $\mathbf{F}=(f_x,f_y)$, luego,

$$
r = \sqrt{x^2+y^2},\\
f_x = F\cos(\theta)=F\frac{x}{r},\\
f_y = F\sen(\theta)=F\frac{y}{r},\\
$$

esto da el sistema de ecuaciones,

$$
\frac{d^2x}{dt^2} = -GM\frac{x}{r^3},\quad\text{y}\quad\frac{d^2y}{dt^2} = -GM\frac{y}{r^3}.
$$

que se puede reescribir  como el sistema de ecuaciones lineales de primer orden, 

$$
\begin{eqnarray}
\frac{dx}{dt}   &=& v_x            &&    x(t+h)=x(t)+ hv_x(t),\\ 
\frac{dy}{dt}   &=& v_y            &&    y(t+h)=y(t)+ hv_y(t),\\
\frac{dv_x}{dt} &=& -GM\frac{x}{r^3} &\overset{\quad\text{ Euler }\quad}{\Longleftrightarrow} &v_x(t+h)=v_x(t)-hGM\frac{x(t)}{r^3}, \\
\frac{dv_y}{dt} &=& -GM\frac{y}{r^3} && v_y(t+h)=v_y(t)-hGM\frac{y(t)}{r^3},
\end{eqnarray}
$$

finalmente se define, $y_{0}=x$, $y_{1}=y$, $y_{2}=v_x,$ y $y_{3}=v_y$, que tiene la forma vectorial $\mathbf{y}'=\mathbf{f}(t,\mathbf{y})$ y resuelve por método de euler.

## 3) Métodos  de verlet y leapfrog para ecuaciones de Newton
En 1967 Verlet propuso un método de segundo order basado en la expansión de taylor y se conoce como *método de verlet*. El
método de verlet para integración de las ecuaciones de Newton  se obtiene a partir de las expansiones,

$$
\begin{align}
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac{a(t)\Delta t^2}{2}+\frac{b(t)\Delta t^3}{6}+O(\Delta t^4)\\
x(t-\Delta t) &=x(t)-v(t)\Delta t+\frac{a(t)\Delta t^2}{2}-\frac{b(t)\Delta t^3}{6}+O(\Delta t^4), 
\end{align}
$$

al sumar da,

$$x(t+\Delta t) = 2x (t) - x (t-\Delta t) + a (t) \Delta t^2 + O(\Delta t^4),$$

y la velocidad se obtiene por diferencia central,

$$v(t) =\frac {x(t+\Delta t)-x(t-\Delta t)}{ 2 \Delta t },$$

dado que en el primer paso de integración ($i=0$) no se conoce $x(-\Delta t)$, entonces se usa la aproximación $x_1=x_0+v_0\,\Delta t+\frac12 a(x_0)\,\Delta t^2$, donde  $(x_0,v_0)$, son las condiciones iniciales y para los demás pasos ($i=1,2,\dots$) se usa las dos formulas anteriores (por simetría también se puede hacer $x(-\Delta t)=x_0+v_0\,(-\Delta t)+\frac12 a(x_0)\,\Delta t^2$ y comenzar $i$ desde cero, $i=0,1,2,\dots$).
Es de notar que aunque el error local en la posición es de orden 4, en las velocidades es de orden 2, el error general es de orden 2. Además la velocidad y la posición se computan a tiempos diferentes, lo cual no es bueno si se quiere computar la energía $E=K+V$. 
Una mejor aproximación es usar el método de *Leap-frog* (que significa salto de rana), este algoritmo se define por,

$$
\begin{align}
v(t + \Delta t/2) &= v(t) + a(t)(\Delta t/2)\\
x(t + \Delta t)   &= x(t) + v(t + \Delta t/2)\Delta t\\
v(t + \Delta t)   &= v(t + \Delta t/2) + a(t + \Delta t)(\Delta t/2)
\end{align}
$$

Otra forma de escribir el método de leapfrog es usar la primera ecuación para remplazar $v(t + \Delta t/2)$ en la segunda y tercera, lo que da la forma más compacta,

$$
\begin{align}
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac {1}{2}a(t)\Delta t^2,\\ 
v(t+\Delta t)&=v(t)+\frac {a(t)+a(t+\Delta t)}{2}\Delta t.
\end{align}
$$

esta forma se conoce cómo *velocidad de verlet* ("velocity Verlet" en inglés). Note que estas dos ecuaciones se pueden también obtener, la primera al hacer una expansión de Taylor hasta el segundo orden en $x(t+\Delta t)$ y la segunda para $v(t+\Delta t)$ por método integración trapezoidal (recordar: $v_b-v_a =\int_a^b f(t)\,dt\approx (f(b)+f(a))\Delta t/2$). 

En general, si se considera el cálculo de las fuerzas, $a(t)=F(x(t))/m$ y $ a(t+\Delta t)=F(x(t+\Delta t))/m$, entonces

$$
\begin{align}
         a(t)&=F(x(t))/m\\
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac {1}{2}a(t)\Delta t^2\\ 
a(t+\Delta t)&=F(x(t+\Delta t))/m\\
v(t+\Delta t)&=v(t)+\frac {a(t)+a(t+\Delta t)}{2}\Delta t,
\end{align}
$$

o cambiando la representación por indices

$$
\boxed{
\begin{align}
  a_i     &= F_i/m\\
  x_{i+1} &= x_i + v_i\, \Delta t + \tfrac{1}{2}\,a_i\, \Delta t^{\,2}, \\
  a_{i+i} &= F_{i+1}/m\\
  v_{i+1} &= v_i + \tfrac{1}{2}(a_i + a_{i+1})\,\Delta t.
\end{align}
}
$$

Para $i=0,1,2,\dots,N$ pasos de integración and $t=i \Delta t$. Los algoritmos de verlet en todas sus formas son reversibles en el tiempo, es decir si se cambian $t$ por $-t$, se recuperan las condiciones iniciales, esto es de gran importancia para el cálculo de la energía, los algoritmos de integración que conservan la energía se conocen como *métodos simplécticos*.

## 4) Método de Runge-Kutta  de segundo orden (rk2)
Considere el problema de valor inicial,

$$ \frac{dy}{dt}=y'=f(t,y),\quad\hbox{ con }\quad a\leq t\leq b, \quad y(a) = \alpha, $$

que se puede solucionar mediante el algoritmo de Runge-Kutta-2 con error de truncación $O(h^2)$ (recuerde que euler es de orden $O(h)$). Para deducir el método rk2 primero se integra la ecuación anterior en el intervalo $[t,t+h]$, así,

$$y(t+h)=y(t)+\int_t^{t+h}f(t',y(t'))dt'$$

Luego se hace expansión de taylor alrededor del punto medio $t_{h/2}=t+h/2$, 

$$f(t',y(t'))=f(t_{h/2},y(t_{h/2}))+(t'-t_{h/2})\frac{d}{dt}f(t_{h/2},y(t_{h/2}))+O(h^2),$$

La integral del primer término es trivial y da $hf(t_{h/2},y(t_{h/2}))$ y la integral del segundo término da cero como se puede ver (pues es función impar y el intervalo de integración  es simétrico respecto al punto medio 
$t_{h/2}$),

$$
\begin{align}
\int_t^{t+h} (t'-t_{h/2})\,dt'
  =&\,\frac{1}{2}(t'-t_{h/2})^2\Big|_t^{t+h} \\
  =&\,\frac{1}{2}\left(t+h-t-\frac{h}{2}\right)^2-\frac{1}{2}\left(t-t-\frac{h}{2}\right)^2\\
  =&\,0.
\end{align}
$$

Así pues obtenemos

$$y(t+h)=y(t)+hf(t_{h/2},y(t_{h/2}))$$

para conocer $y(t_{h/2})$ hacemos expansión alredor de $t$ al primer orden, lo que da 

$$y(t_{h/2})\approx y(t)+(t_{h/2}-t)\frac{d}{dt}y(t)=y(t)+\frac{h}{2}f(t,y(t)).$$

donde se ha usado que $y'=f(t,y(t))$ y $(t_{h/2}-t)=h/2$, finalmente si definimos las constantes $k_1$ y $k_2$ se tiene el algoritmo de rk2,

\begin{align}
k_1   =&\, hf(t,y)\\
k_2   =&\, hf\left(t+\frac{h}{2},y+\frac{k_1}{2}\right)\\
y(t+h)=&\, y(t)+k_2.
\end{align}

Esta forma particular se conoce como método de punto medio. 
Implementación:

````python
def rk2( f, y, t, h ):

    k1 = h*f(t, y)
    k2 = h*f(t + 0.5*h, y + 0.5*k1)
    y = y + k2

    return y
````

### Sistemas de ecuaciones diferenciales de primer orden
Considere que que se tiene un sistema de $N$ partículas que interactúan entre sí, cada partícula sigue las leyes de Newton por lo que se tendrá a partir de la segunda ley un sistema de $3N$ ecuaciones acopadas, esto constituye lo que en física se conoce como el problema de $N$ cuerpos y desafortunadamente solo tiene solución para $N\le2$, así que se deben implementar métodos numéricos para resolver este problema si $N\ge3$.
Runge-Kutta se puede usar para resolver sistemas de ecuaciones, para esto primero las ecuaciones se deben poner en forma vectorial lineal de orden uno,  

$$
\begin{pmatrix}
  \frac{d}{dt}y_1(t)\\
  \frac{d}{dt}y_2(t)\\
  \vdots\\
  \frac{d}{dt}y_n(t)\\
\end{pmatrix}
=
\begin{pmatrix}
  f_1(t,y_1,y_2,\cdots,y_n)\\[7pt]
  f_2(t,y_1,y_2,\cdots,y_n)\\
  \vdots\\
  f_n(t,y_1,y_2,\cdots,y_n)\\
\end{pmatrix}
\quad\Longrightarrow\quad
\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}),
$$

con $a\leq t\leq b,$ y $\mathbf{y}(a) = \boldsymbol{\alpha}$, que es la forma deseada para resolver numéricamente con runge-kutta.

### Ecuaciones diferenciales de orden superior
El orden de una ecuación diferencial es dado por la derivada de más alto orden, así por ejemplo las ecuaciones de Newton son de segundo orden, en general una ecuación lineal de orden $n$ tiene la forma,

$$a_n(t)\frac{d^ {n}}{dt^n}y(t)+a_{n-1}(t)\frac{d^{n-1}}{dt^{n-1}}y(t)+\cdots+a_1(t)\frac{d}{dt}y(t)+a_0(t)y(t)=f(t),$$

donde las $a_i$ son funciones de $t$, esta ecuación se puede transformar a un sistema lineal de $n$ ecuaciones de primer orden, si se realizan las siguientes transformaciones: se define $y=y_0$ y una nueva variable $y_i$ por cada derivada, esto transforma la ecuación de orden $n$ a un sistema de $n$ ecuaciones acopladas de primer orden. Supongamos por simplicidad que, $a_n(t)=1$, entonces esto da el sistema,

$$
\begin{align}
  \frac{d}{dt}y_0(t)=&\, y_1,\\
  \frac{d}{dt}y_1(t)=&\, y_2,\\
  \vdots\\
  \frac{d}{dt}y_{n-1}(t)=&\, f(t)-\left[a_{n-1}(t)y_{n-1} +\cdots +a_1(t)y_1 +a_0(t)y_0\right],   
\end{align}
$$

donde las $n$ variables, $y_0,\cdots,y_{n-1}$, se numeran de 0 a $n-1$ para ser más amigable con la notación de arrays de python.

**Ejemplo** consideremos la ecuación de orden 3,

$$t^2\frac{d^ {3}}{dt^3}y(t)+\,\hbox{sen}(t+2)\frac{d^{2}}{dt^{2}}y(t)+e^{2t}\frac{d}{dt}y(t)+\cos(t)=0,$$

Si se divide la ecuación por $t^2$ se tiene que $a_2(t)=t^{-2}\,\hbox{sen}(t+2)$, $a_1(t)=t^{-2}e^{2t}$, $a_0(t)=0$, y $f(t)=-t^{-2}\cos(t)$, que da el sistema

$$
\begin{align}
  \frac{d}{dt}y_0(t)  =&\, y_1,\\
  \frac{d}{dt}y_1(t)  =&\, y_2=\,\frac{d}{dt}\left(\frac{d}{dt}y_0(t)\right),\\
  \frac{d}{dt}y_{2}(t)=&\, f(t) -a_{2}(t)y_{2} -a_1(t)y_1 .   
\end{align}
$$

Remplazando en forma vectorial da,

$$
\begin{pmatrix}
  \frac{d}{dt}y_0(t)\\
  \frac{d}{dt}y_1(t)\\
  \frac{d}{dt}y_2(t)\\
\end{pmatrix}
=
\begin{pmatrix}
\, y_1\\[5pt]
\, y_2\\
\, -t^{-2}(\,\hbox{sen}(t+2)y_{2} +e^{2t}y_1 +\cos(t))\\
\end{pmatrix}
$$

en el cual reconocemos el sistema $\mathbf{y}' =\mathbf{f}(t,\mathbf{y})$ que se puede resolver con Runge-Kutta. Note que para el caso del sistema de $N$ ecuaciones de Newton cada ecuación debe primero ser puesta como un sistema de dos ecuaciones de primer orden, es decir, una EDO para la velocidad y otra EDO para la posición, esto por cada dimensión y como el espacio es dimensión tres, el problema de $N$ cuerpos da un sistema de $6N$ ecuaciones de primer orden a resolver.

## 5) Método de Runge-Kutta de cuarto orden (rk4)
Runge-Kutta-2 no se utiliza casi y solo se muestra por didáctica, en su lugar se usan métodos de más alto orden. Runge-Kutta-4 (rk4) es uno de los métodos más populares y tiene error de truncación numérica de orden $O(h^4)$, la deducción es similar a rk2 pero mucho mas larga y se omite ([ver deducción en wikipedia](https://en.wikipedia.org/wiki/Runge%E2%80%93Kutta_methods#Derivation_of_the_Runge.E2.80.93Kutta_fourth-order_method)). 
Considere el sistema de $N$ ecuaciones acopladas (en negrilla para recordar el caracter vectorial, ver ejemplo abajo),

$$\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}) ,\\ \\ \\ a\leq t\leq b, \\ \\ \\ \\ \mathbf{y}(a) = \boldsymbol{\alpha},$$

donde $\mathbf{y}(t)=\{y_1(t),y_2(t),\cdots,y_N(t)\}$ son la variables a integrar y $\mathbf{f}(t,\mathbf{y})=\{f_1,f_2,\cdots,f_N\}$.
El método iterativo rk4 usa las siguientes expresiones para resolver este sistema

$$ 
\begin{align}
 \mathbf{k}_1 =&\, h\mathbf{f}\,(t,\mathbf{y})\\
 \mathbf{k}_2 =&\, h\mathbf{f}\left( t + \frac{h}{2},\mathbf{y} + \frac{\mathbf{k}_1}{2}\right)\\
 \mathbf{k}_3 =&\, h\mathbf{f}\left( t + \frac{h}{2},\mathbf{y} + \frac{\mathbf{k}_2}{2}\right)\\
 \mathbf{k}_4 =&\, h\mathbf{f}\left( t + h,\mathbf{y} + \mathbf{k}_3\right)\\
 \mathbf{y}(t+h) =\,& \mathbf{y}(t) + \frac{1}{6}( \mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4  )
 \end{align}
$$

Implementación en python:

````python
def rk4(f,y,a,b,n):
    '''------------------ Runge-kutta 4  ------------------------
    resuelve un sistema con varias ecuaciones diferenciales (ode)
    '''
    t = a
    d  = len(y)      # Obtener número de ecuaciones.
    k1 = k2 = k3 = k4 = zeros(d)
    
    tt = zeros(n)    # Array tiempo.
    Y  = zeros((d,n))# Array coords.
    h = (b - t)/n
    ti = 0
    while ti<n:

        Y[:, ti] = y # Almacene como columna coords al tiempo t_i
        tt[ti] = t   # Almacene el tiempo.

        # *** Runge kutta 4 método ***   
        k1 = h*f(t, y)
        k2 = h*f(t + h/2., y + k1/2.)
        k3 = h*f(t + h/2., y + k2/2.)
        k4 = h*f(t + h, y + k3)
        
        y = y + (k1 + 2.*(k2 + k3) + k4)/6.
    
        t = t + h   # Incremente el tiempo por h.
        ti = ti + 1 # Incremente el contador.  
        
    return Y,tt     # Output: arrays con coords y tiempo para graficar.
````

**Ejemplo**: En las siguientes lineas de python se usa rk4 para resolver el oscilador armónico forzado, 

$$ \frac{dy^2}{dt^2}= -100y -2\frac{dy}{dt} + 10\sen(3t),$$

con condiciones iniciales $y_0 = 3$ y   $v_0 = -5$.
Si se define $v = \frac{dy}{dt}$, da el sistema vectorial

\begin{align}
\frac{dy}{dt} = &\,\, v\\
\frac{dv}{dt} = &\,\, -100y -2v + 10\sen(3t)
\end{align}

En el siguiente código el vector $(y,v)$ se representa con el array $(y[0],y[1])$ y la función $\mathbf{f}=(f_1,f_2)$:

````python
# Usar Runge Kutta 4 (rk4) 
from numpy import *
import matplotlib.pyplot as plt
%matplotlib inline

a, b = 0, 10.  # intervalo de integración
n = 10000      # Número max de pasos de integración
y0 = array([3., -5.]) # Condición inicial (pos, vel)

def f(t, y):   # función a integrar de la forma y'(t) = f(t,y(t))       
    f1 = y[1] 
    f2 = -100.*y[0] -2.*y[1] + 10.*sin(3.*t)
    return array([f1, f2])


Y, time = rk4(f,y0, a,b, n)  # llamar rk4
    
plt.plot(time, Y[0])
plt.plot(time, Y[1])
plt.legend( ["$y(t)$", "$v(t)$"], numpoints=1)
plt.xlabel('t')
plt.show()
````

### Estabilidad de ecuaciones diferenciales
Se refiere a que tanto puede cambiar la solución si se hace un pequeño cambio en las condiciones iniciales, "*en sistemas caóticos pequeños cambios pueden dar soluciones totalmente diferentes*".
Para verificar numéricamente que tan sensible es una ecuación (o sistema de ecuaciones) se observa si un 
pequeño cambio en las condiciones iniciales crece exponencialmente, ósea si se considera 
$x_0$ y $y_0=x_0+\epsilon$, entonces se define $|x(t)−y(t)|$ y se mira si esta cantidad crece exponencialmente, si este es el caso el sistema es caótico (más formalmente, para $\epsilon\to0$ y $t\to\infty$ el exponente de la dependencia anterior, si existe, se conoce como exponente de Lyapunov).

### Atractor
un [atractor](https://es.wikipedia.org/wiki/Atractor) es un conjunto de valores numéricos hacia el cual un sistema de ecuaciones tiende a evolucionar, en una amplia variedad de condiciones iniciales del sistema. Por ejemplo los sistemas dinámicos en física que son disipativos (por fricción interna, pérdidas termodinámicas, pérdida de masa etc) tienen comportamiento de atractores.

**Ejercicio** En 1963 Eduard Lorenz desarrolló un modelo matemático para calcular la inestabilidad de la atmósfera, este modelo se conoce como [el  atractor de lorenz](https://en.wikipedia.org/wiki/Lorenz_system#Python_simulation) y se define por la siguientes ecuaciones,

$$
\begin{align}
\frac{\mathrm{d}x}{\mathrm{d}t} &= \sigma (y - x), \\
\frac{\mathrm{d}y}{\mathrm{d}t} &= x (\rho - z) - y, \\
\frac{\mathrm{d}z}{\mathrm{d}t} &= x y - \beta z.
\end{align}
$$
Que tiene comportamiento caótico para ciertos valores de $\sigma, \rho$ y $\beta$.
La siguiente rutina usa la función de la librería scipy `scipy.integrate.odeint` implementada para integrar ecuaciones diferenciales, luego comparamos el resultado con rk4: 

````python
# 1) usa scipy para integrar
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint      # método de integracion scipy
from mpl_toolkits.mplot3d import Axes3D # hacer gráfico 3D

rho = 28.0
sigma = 10.0
beta = 8.0 / 3.0

def f(state, t):
  x, y, z = state             # crear una tupla de la forma (x,y,z)       
  return sigma * (y - x), x * (rho - z) - y, x * y - beta * z  

state0 = [1.0, 1.0, 1.0]      # condición inicial
t = np.arange(0.0, 40.0, 0.01)

states = odeint(f, state0, t) # Integrar el sistema ode, dy/dt = f(y, t0, ...)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(states[:,0], states[:,1], states[:,2])
plt.show()
````

````python
# 2) usa rk4 para integrar
# Inicialización
a, b = 0., 40.         # Intervalo [a,b] 
n = 10000              # max número de pasos de integración
y0 = array([1.,1.,1.]) # condiciones iniciales 

rho = 28.0; sigma = 10.0; beta = 8.0/3.0
fR = zeros(3)          # inicializar array

# función f(t,y) = sigma * (y - x), x * (rho - z) - y, x * y - beta * z
def f(t, y):
    fR[0] = sigma * (y[1] - y[0]) 
    fR[1] = y[0] * (rho - y[2]) - y[1]
    fR[2] = y[0] * y[1] - beta * y[2]
    return fR
    
Y, time = rk4(f,y0, a,b, n)  # integrar
    
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(Y[0], Y[1], Y[2])
ax.set_xlabel('X(t)')
ax.set_ylabel('Y(t)')
ax.set_zlabel('Z(t)')

plt.show()
````

Note que para $\rho = 28.0, \sigma = 10.0$ y $\beta = 8.0/3.0$ el atractor de lorentz produce soluciones totalmente diferentes si se hacen pequeñas variaciones de las condiciones iniciales $(x_0,y_0,z_0)$ (por ejemplo al sumar $10^{-5}$ a $x_0$, para ver esto grafique ambos casos y compare). Otro caso interesante es cuando $\rho<1$, pues todas las trayectorias convergen al origen $(0,0,0)$ (otra vez, grafique dos trayectorias con condiciones iniciales diferentes para ver esto).
El atractor de lorentz es un ejemplo clásico de sistemas caóticos.

### Runge-Kutta-Fehlberg Method (rk45)
 Este es otro método popular que tiene control del error a cada paso, y por lo tanto es de paso, $h_i$, variable. Se llama rk45 debido a que combina una aproximación $O(h^{4})$ de Runge-Kutta junto con otra de orden $O(h^{5})$ para calcular el error, es decir, sabemos que el error absoluto es dado por $\text{err} =|y_{i+1} - y|$, donde $y$ es la solución exacta a $t_i +h$, dado que no se conoce $y$ entonces se aproxima al valor $y\approx z_{i+1}$ de orden  $O(h^{5})$. Osea, la aproximación de Runge-Kutta a $O(h^{4})$ es,



$$ 
 \mathbf{y}_{i+1} = \mathbf{y}_{i}  + \frac{25}{216}\mathbf{k}_1 + \frac{1408}{2565}\mathbf{k}_3 + \frac{2197}{4104}\mathbf{k}_4 - \frac{1}{5}\mathbf{k}_5,
$$

 pero una mejor aproximación de Runge-Kutta a orden $O(h^{5})$ es,
 
$$
\mathbf{z}_{i+1} = \mathbf{y}_{i}  + \frac{16}{135}\mathbf{k}_1 + \frac{6656}{12825}\mathbf{k}_3 + \frac{28561}{56430}\mathbf{k}_4 - \frac{9}{50}\mathbf{k}_5 + \frac{2}{55}\mathbf{k}_6,
$$
 
donde,

$$
\begin{eqnarray}
\mathbf{k}_1 &=& h\mathbf{f}\left(t_i, \mathbf{y}_i\right), \\
\mathbf{k}_2 &=& h\mathbf{f}\left(t_i + \frac{h}{4}, \mathbf{y}_i + \frac{\mathbf{k}_1}{4}\right),\\
\mathbf{k}_3 &=& h\mathbf{f}\left(t_i + \frac{3}{8}h, \mathbf{y}_i + \frac{3}{32}\mathbf{k}_1 + \frac{9}{32}\mathbf{k}_2\right),\\
\mathbf{k}_4 &=& h\mathbf{f}\left(t_i + \frac{12}{13}h, \mathbf{y}_i + \frac{1932}{2197}\mathbf{k}_1 - \frac{7200}{2197}\mathbf{k}_2 + \frac{7296}{2197}\mathbf{k}_3\right),\\
\mathbf{k}_5 &=& h\mathbf{f}\left(t_i + h, \mathbf{y}_i + \frac{439}{216}\mathbf{k}_1 - 8\mathbf{k}_2 + \frac{3680}{513}\mathbf{k}_3 - \frac{845}{4104}\mathbf{k}_4\right),\\
\mathbf{k}_6 &=& h\mathbf{f}\left(t_i + \frac{1}{2}h, \mathbf{y}_i - \frac{8}{27}\mathbf{k}_1 + 2\mathbf{k}_2 - \frac{3544}{2565}\mathbf{k}_3 + \frac{1859}{4104}\mathbf{k}_4 - \frac{11}{40}\mathbf{k}_5\right),
\end{eqnarray}
$$

el paso más óptimo es dado con $h\rightarrow sh$,

$$
s=\left(\frac{\epsilon h}{2\|\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\|}\right)^{\frac{1}{4}}\approx 
0.84\left(\frac{\epsilon h}{\text{err}}\right)^{\frac{1}{4}},
$$

donde el error entre las dos aproximaciones es dado por, $\text{err} =\|\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\|$, pero por eficiencia computacional es mejor calcularlo como,

$$
\begin{eqnarray}
\text{err} 
&=& \text{max}\left(\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\right),\\
&=& \text{max}\left(\frac{\mathbf{k}_1}{360} - \frac{128}{4275}\mathbf{k}_3 - \frac{2197}{75240}\mathbf{k}_4 + \frac{\mathbf{k}_5}{50} + \frac{2}{55}\mathbf{k}_6\right),
\end{eqnarray}
$$        

donde $\text{max}()$ retorna la componente vectorial más grande en valor absoluto. Note que para determinar el $h$ más optimo a cada paso se usan dos diferentes aproximaciones de la solución, $\mathbf{y}_{i+1}$ y $\mathbf{z}_{i+1}$, y se comparan, si las dos aproximaciones son similares de acuerdo a cierta tolerancia $\epsilon$ se acepta $\mathbf{y}_{i+1}$ y se incrementa o reduce $h$ multiplicando por $s$, sino, igualmente se escala $h$ a $sh$ y se recalcula $\mathbf{y}_{i+1}$. Notemos que si `err` es pequeño $s$ será grande y $h$ incrementará pero si `err` es grande $s$ será pequeño $h$ disminuirá.
Por estabilidad y eficiencia los valores de $s$ y $h$ se restringen a los intervalos $0.1 \leq s \leq 4.0$ y  $h_{min} \leq h \leq h_{max}$.
Aunque rk45 calcula 6 veces la función $f(t,y)$ es en general más rápido y preciso que rk4 (para más detalles ver el libro de Burden). La implementación de rk45 con $\epsilon = 10^{-5}$ es :

````python
def rk45(f, y,t,b, hmin = 0.001, hmax = 2., Tol = 1e-5):     
    '''-------------------- Runge-Kutta-Fehlberg (rk45) ------------------------
    Runge-Kutta-Fehlberg método, integra el sistema vectorial, 
                     y'(y) = f(t,y(t)) 
    f: función vectorial, y: condición inicial, (t,b): intervalo de integración, 
    (hmin, hmax): valores minimo y máximo de h, Tol: tolerancia en el error. 
    '''

    d  = len(y)         # tamaño del sistema de ecuaciones 
    k1 = k2 = k3 = k4 = k5 = k6 = np.zeros(d)
    
    h = hmax            # inicializar h al valor máximo 
    tt = t              # array 1 dim del tiempo
    Y  = np.array([y])  # array 2 dim con coords      

    while t < b:
        # *** Runge Kutta Fehlberg método (rk45) ***   
        k1 = h*f(t, y)                
        k2 = h*f(t + h/4, y + k1/4)
        k3 = h*f(t + 3/8*h, y + 3/32*k1 + 9/32*k2)
        k4 = h*f(t + 12/13*h, y + 1932/2197*k1 - 7200/2197*k2 + 7296/2197*k3)
        k5 = h*f(t + h, y + 439/216*k1 - 8*k2 + 3680/513*k3 - 845/4104*k4)
        k6 = h*f(t + 1/2*h, y - 8/27*k1 + 2*k2 - 3544/2565*k3 + 1859/4104*k4 - 11/40*k5)
        # Calcular el error |z(i+1) − y(i+1)|
        err = (np.abs(k1/360 - 128/4275*k3 - 2197/75240*k4 + k5/50 + 2/55*k6)).max()
        # Aceptar or rechazar el paso de integración i+1 
        if (err/h < Tol): 
  
           y = y + 25/216*k1 + 1408/2565*k3 + 2197/4104*k4 - k5/5
           t = t + h               # incremente en h el tiempo

           Y  = np.append(Y, np.array([y]), axis=0)# almacenar coords al tiempo t_i
           tt = np.append(tt,t)                    # almacenar tiempo en el array
           
        # calcular el valor de h para el siguiente paso, verificar que h no sea muy 
        # grande o muy pequeño.  
        s = 0.84*(Tol*h/err)**0.25 # scalar a multiplicar h--> s*h
        
        if (s <= 0.1): h = 0.1*h   # (err grande) 0.1 es el min valor permitido de s
        elif (s >= 4): h = 4.*h    # 4 es el max valor permitido de s
        else:          h = s*h     # escalar h a s*h, paso ideal ( 0.1 < s < 4)
        
        if   (h > hmax): h = hmax  # no permitir h más grande que hmax
        elif (h < hmin): 
            print('(Minimo h excedido, Procedimiento completado sin éxito.')
            break
            
        if (t+h > b): h = b - t    # garantizar que en el último paso t = b  
                    
    return Y.T, tt # output: arrays con coords y tiempo para gráficar.
````

**Ejemplo**: Comparación de rk4 and rk45 para el oscilador armónico forzado anterior.

````python
import numpy as np
import matplotlib.pyplot as plt
#%matplotlib qt       # abrir gráfico en ventana exterior

a, b = 0, 10.         # intervalo de integración
n = 10000             # Número max de pasos de integración
y0 = np.array([3., -5.]) # Condición inicial (pos, vel)

def f(t, y):          # función a integrar de la forma y'(t) = f(t,y(t))       
    f1 = y[1] 
    f2 = -100.*y[0] -2.*y[1] + 10.*np.sin(3.*t)
    return np.array([f1, f2])

# 1)--------------- plot rk45 -------------------------
Y, t = rk45(f, y0, a, b)          # llamar rk45

print('Max time {} reached at step {} of {} max steps allowed.'.format(t[-1],len(t),n) ) 

plt.plot(Y[0], Y[1],label='rk45') # note que Y[i] es igual a Y[i,:]
plt.plot(Y[0,-1],Y[1,-1],'*')     # mostrar punto al final de rk45.

#2) --------------- plot rk4  -------------------------
Y1, t1 = rk4(f, y0, a, b, n)      # llamar rk4

plt.plot(Y1[0], Y1[1],label='rk4',c='k')
plt.plot(Y1[0,-1],Y1[1,-1],'*')   # mostrar punto al final.

plt.title('Phase space')
plt.ylabel('$v(t)$')
plt.xlabel('$x(t)$')
plt.legend()
plt.show()    
````

````python
# Lorentz atractor with rk45
a, b = 0., 40.         # Intervalo [a,b] 
y0 = array([1.,1.,1.]) # condiciones iniciales 

rho = 28.0; sigma = 10.0; beta = 8.0/3.0
fR = zeros(3)          # inicializar array

# funcion f(t,y) = sigma * (y - x), x * (rho - z) - y, x * y - beta * z
def f(t, y):
    fR[0] = sigma * (y[1] - y[0]) 
    fR[1] = y[0] * (rho - y[2]) - y[1]
    fR[2] = y[0] * y[1] - beta * y[2]
    return fR
    
Y, time = rk45(f,y0, a,b)  # integrar
print('Max time {} reached at step {}.'.format(time[-1],len(time))) 
    
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(Y[0], Y[1], Y[2])
ax.set_xlabel('X(t)')
ax.set_ylabel('Y(t)')
ax.set_zlabel('Z(t)')

plt.show()
````
