# Representación binaria
Los números y caracteres se representan en computación con ceros y unos. Un cero o un uno se conoce como "bit", estos a su vez se agrupan en paquetes de 8 bits que son un byte.
La memoria se mide en bytes:

* kilobyte, 1KB
* megabyte, 1MB
* gigabyte, 1GB

Debido a que se usa base 2 y no 10 un KB es 1024 bytes (no bits), entonces 1MB es 1024*1024 bytes y 1GB es 1024$^3$ bytes.

La representación numérica depende del número de caracteres o símbolos usados para representar un número, en base decimal se usan 10 caracteres, en base 2 se usan solo 2 y en base hexadecimal se usan 16 símbolos. En la siguiente tabla se muestra una comparación de las tres bases,

 Note que con 4 bits solo se puede representar máximo 16 números (del 0 al 15), en general, con $N$ bits se tienen $2^N$ combinaciones y el entero más grande es $2^N-1$. Si se usan 8 bits da un máximo de $2^8=256$ diferentes combinaciones que se usan para representar los 256 caracteres del código [ASCII](https://es.wikipedia.org/wiki/ASCII) que es basado en el alfabeto latino, así por ejemplo el carácter `a`  es representado por 01100001 y `A` por 01000001 (las mayúsculas y minúsculas difieren solo por un bit, lo cual facilita la conversión). En el código ASCII los dígitos del 0 al 9 se representan también con 8 bits (para saber su representación, de la tabla anterior se prefija el valor 0011 a cada uno de los binarios del 0 al 9, por ejemplo el cero tiene el código ASCII 00110000). Así por ejemplo un archivo que solo contenga la palabra "hola" tiene un tamaño de 4 bytes o 32 bits. 

### Representación binaria de enteros
Para pasar de base 10 a binarios se divide por 2 hasta llegar a 1, ejemplo
```c
       445/2      1 
       222/2      0 
       111/2      1
        55/2      1
        27/2      1
        13/2      1
         6/2      0
         3/2      1
         1/2      1
```  
Entonces el binario es 110111101. Verifiquemos que en verdad este binario es 445:

$$1\times2^8 + 1\times2^7 + 0\times2^6 + 1\times2^5 + 1\times2^4 + 1\times2^3 + 1\times2^2 + 0\times2^1 +1\times2^0=445.$$

Un número entero en python3 es tipo `int` y se implementa a bajo nivel mediante el tipo `long` de C. 
Los enteros en python3 son de precisión arbitraria (un entero tan grande como se quiera), como se puede ver en ipython usando la siguiente rutina
```python
    long=1
    for n in range(1000):
        long=long*2
        print(n, long) 
```       
en python 2.7 el mismo bucle también da un número de 302 cifras (con una `L` al final, ver más abajo), ya que python 2.7 en realidad tiene dos tipos enteros distintos: `int`, que corresponde directamente al `long` de C, acotado a 32 (o 64) bits, y `long`, que sí es de precisión arbitraria y es al que python promueve automáticamente un `int` cuando se desborda. El tipo `int` de C con 32 bits da un rango $-2^{31}< x < 2^{31} – 1$, (o  $-2147483648$ a $2147483647$, es decir con 10 cifras significativas); note que el exponente es 31 porque un bit se reserva para guardar el signo. Con 64 bits, el rango es $-9223372036854775808 < x< 9223372036854775807$, es decir con 19 cifras significativas. En python 3.x esta distinción desaparece: solo existe `int`, que siempre es de precisión arbitraria (en python 2.7, a los números de tipo `long` se les agrega una `L` al final para distinguirlos de `int`; esta notación ya es obsoleta en python 3.x).

### Representación binaria de números de punto flotante
Los números con punto decimal se conocen como números de punto flotante, y en los 
computadores modernos se representan con 32 bits (tipo float) o 64 bits (tipo double). (NOTA: la palabra *float* tiene 2 significados según el contexto, números con punto decimal y también representación de 32 bits).

La forma más sencilla para convertir un número de punto decimal a binario es separando la parte entera y la fraccional. 
En la parte entera se divide por 2 y el residuo da los bits, en la parte fraccional se multiplica por 2, el resultado se separa otra vez en la parte entera y fraccional, la parte entera será 0 o 1, esta nos da los bits, la fraccional se vuelve a dividir hasta que nos de 1.0.

Ejemplo, el número -1313.3125 se pasa a binario así:
```c
        parte entera                     parte fraccional
     División     Residuo          Fracción x 2          Entero
                           
     1313/2       1                0.3125 × 2 = 0.625    0 
      656/2       0                0.625  × 2 = 1.25     1 
      328/2       0                0.25   × 2 = 0.5      0 
      164/2       0                0.5    × 2 = 1.0      1
       82/2       0             
       41/2       1             
       20/2       0             
       10/2       0             
        5/2       1             
        2/2       0             
        1/2       1             
```
El resultado da para la parte entera $10100100001_{2}$ y la fraccional $0101_{2}$. Si sumamos las dos partes el resultado es $-10100100001.0101_{2}$. Para convertir de base 2 a base 10 se hace con potencias negativas de 2,

$$0\times2^{-1}+ 1\times2^{-2} + 0\times2^{-3} + 1\times2^{-4} =0.3125$$

### Truncamiento
Los números reales no siempre se pueden representar en el computador debido a que muchos de ellos requieren una representación infinita en base 2 (esto significa una memoria infinita para almacenar el número), tal es el caso por ejemplo de los números  $\pi$, $e$, 1/10 o 1/3, pues,
```c
    pi   = 3.1416... = 11.001001000011111101101010100010001000010110...
    e    = 2.7182... = 10.101101111110000101010001011000101000101011...
    1/10 = 0.1       =  0.000110011001100110011001100110011001100110...
    1/3  = 0.3333... =  0.01010101010101...
```    
son infinitos en base 2. Esto significa que solo se toman un número determinado de bits para representar el número, esto se conoce como error de truncamiento, pues el número truncado es una aproximación al número real. Note que esto no es exclusivo de la base 2, en base 10 el número 1/10 tiene representación exacta 0.1, pero 1/3 no tiene un representación finita, así, se puede representar con 0.33 o con 0.333333333 que es mejor pero no es exactamente 1/3. Esto tiene efecto en las operaciones aritméticas y por lo tanto 0.1\*10 no es exactamente 1 sino una aproximación, similarmente en base 10, pues 3\*0.33333333 = 0.99999999 y no 1. 

### Notación exponencial y normalización
La notación exponencial también se puede usar en base $2$ simplemente corriendo el punto como se hace en notación decimal, por ejemplo, el anterior número $-1313.3125_{10}$ se puede escribir como,

$$-10100100001.0101_{2}=-1.01001000010101_{2}\times2^{10},$$

en especial, si se corre el punto hasta el primer 1, un bit antes de la izquierda, a esto se conoce como *normalización*.

````python
# Convertir flotante en base 10 a binario ()
# Nota: para enteros mejor usar la función bin() de python.

def ftob(x, Tol=20):   # Tol: max num de dígitos después del pto
    
    sig = "-" if x<0 else "" # signo del numero
    n = int(x)         # tomar parte entera del x
    f = abs(x - n)     # tomar parte fraccional de x
    
    # *** Calculo parte entera (dividir por 2) 
    L = ""             # crear string parte entera
    while abs(n)>0:
        L = str(n%2)+L # residuo de dividir por 2    
        n = int(n/2)   # parte entera de dividir por 2
            
    # *** Calculo parte fracional (multiplicar por 2)
    S = ""             # crear string parte fracional
    dot = "." if f != 0 else ""
    while f != 0:
        D = f*2        # multiplicar fracional por 2 
        i = int(D)     # tomar parte entera (es 0 o 1)
        S = S + str(i) # almacenar en string el binario
        f = D - i      # remover parte entera
        if len(S) > Tol: break # truncar si "max num de dígitos" > Tol
    
    return sig+L+dot+S # retorna string con unos y ceros

print(ftob(-1313))
print(ftob(-1313.3125))    
print(ftob(513.01,100))
# ftob(3.1415926)
````

````python
# Convertir "binario" (debe ser string) a base 10
def btof(x):
    if isinstance(x,str):# verificar si x es string
       p = x.split('.')  # separar parte entera y fraccional
       n = list(p[0])    # crear lista parte entera binaria
       n = n[::-1]       # invertir orden de la lista
       
       # crear parte entera d y fraccional s en base 10:
       d = sum([ int(n[i])*2**i for i in range(len(n))])
       num = d
    
       # crear parte entera d y fraccional s en base 10:
       if len(p)>1:      # verificar si hay parte fraccional 
          f = list(p[1]) # crear lista parte fraccional binaria
          s = sum([ int(f[i])*2**(-i-1) for i in range(len(f))]) 
          num = d+s      # agregar parte fraccional
        
       return num        # retorna numero flotante en base 10
    else:
       print("El binario debe ser string, e.g., '10.11'.") 

print(btof('10100100001.0101')) # 1313.3125
print(btof('1000000001'))       # 513
print(btof('1000000001.000000101000111101011'))
print(btof('1000000001.000000101000111101011100001010001111010111'))
print(btof('11.001001000011111101101010100010001000010110'))
btof(ftob(3.1415926)) # o también así
````

## Conversión de binarios a hexadecimales y viceversa
Al usar la [tabla anterior](#Conversiones) vemos que cada carácter se cambia por su representación binaria de 4 bits, por ejemplo,
```c
     1E3 = 0001 1110 0011
    0A2B = 0000 1010 0010 1011
    7E0C = 0111 1110 0000 1100
```     
Ahora considere el número binario $1000011101_2$, por la misma regla su representación hexadecimal es, 
```c
     0010 0001 1101 = 21D
```
Note que se agregaron dos ceros al principio con el fin de completar el primer digito hexadecimal, pues usualmente los ceros a la izquierda se ignoran. 

Por último, python tiene las funciones `bin(), oct()` y `hex()` para convertir enteros a binario, octal, y hexadecimal, por ejemplo, la representación del 7 en las 3 bases es:

Note que `0b`, `0o` y `0x` al comienzo de cada número significa binario, octal y hexadecimal. Para conversión de floats base 10 a hexadecimal python ofrece la función `float.hex()`.

````python
# representación del 7 en las 3 bases:
print( bin(7), oct(7), hex(7) ) # imprime  0b111 0o7 0x7
print(0b0111, 0o7, 0x07)        # imprime 7 7 7
````

## Convención IEEE 754
La representación binaria anterior no es la más adecuada en el computador, el cual solo ve ceros y unos y no el signo menos o el punto entre la parte fraccional y la entera. los computadores modernos usan la convención IEEE 754, en esta convención un número se representa con 16, 32 o 64 bits, veamos los casos de 32 y 64 bits.

**Representación con 32 bits**: en este caso el primer bit $b_{31}$ es el signo, los siguientes 8 bits $\{b_{30}b_{29} \dots b_{23}\}$ son el exponente y los 23 restantes son la *base* o *mantisa* escrita en forma normalizada (es decir, la parte entera siempre es 1., mas la *parte fraccional* $f$; representada por los restantes bits), ósea:

$$(−1)^s(1 + f)2^{c−127},$$

o más explicitamente, si el exponente en base dos es $c=(b_{30}b_{29} \dots b_{23})_2$ y la mantiza $f=(b_{22}b_{21} \dots b_0)_2$, entonces,

$$(-1)^{b_{31}} \times (1.b_{22}b_{21} \dots b_0)_2 \times 2^{(b_{30}b_{29} \dots b_{23})_2 - 127},$$

el cambio a base 10 es simplemente,

$$(-1)^{b_{31}}\times \left(1 + \sum_{i=1}^{23} b_{23-i} 2^{-i} \right)\times 2^{(c-127)}.$$

Note que la mantisa está normalizada (por eso el "$1.$" seguido de los 23 bits que representan la fracción $f$), ademas, note que el exponente dado por $c$ es un número positivo entre 0 y 255 en base 10, para garantizar números positivos y negativos en el exponente, entonces se le resta 127. En la memoria los bits se almacenan de la siguiente manera,

    * Signo : 1 bit (el primero)
    * Exponente: 8 bits (los 8 siguientes)
    * Mantisa: 24 bits (23 explicitamente almacenados)
  

<img src="../figures/Float_example_1.png" alt="Drawing" style="width: 1000px;"/>

La representación de 32 bits en base 10 tiene aproximadamente 7 cifras significativas, (pues $2^{24}-1= 16777215$ es la mantisa más grande que se puede almacenar, ¿y cuál es el exponente más grande que se puede almacenar?).

**Ejemplo**: en el ejemplo anterior tenemos que $-1313.3125_{10}=-10100100001.0101_2$, expresemos el resultado en notación IEEE 754 que es más practica:

1) Normalizamos como se explicó anteriormente: $10100100001.0101 = 1.01001000010101 \times 2^{10}$

2) la parte fraccional de la mantisa es $01001000010101000000000$, (note que se agregan los ceros restantes a la derecha) y el exponente es $c = 10 + 127 = 137 = 10001001$, 

3) El primer bit debe ser 1 pues el número es negativo. 

Finalmente $-1313.3125$ es 1 10001001 01001000010101000000000.

**Representación con 64 bits**: en el estándar IEEE 754 un "double" se especifica por un binario de 64 bits de la siguiente manera (cada espacio ocupa un bit $b_i$ en la posición $i$):

    * Signo : 1 bit (el primero)
    * Exponente: 11 bits (los 11 siguientes)
    * Mantisa: 53 bits (52 explicitamente almacenados)

<img src="../figures/Double_example_1.png" alt="Drawing" style="width: 1000px;"/>

Esto da que el número en base 10 que tiene 17 cifras significativas de precisión, o equivalentemente 16 cifras despues del punto si se usa notación científica (¿cómo se sabe esto? simplemente mirando $2^{53}$ en base 10). En python los flotantes son por defecto de 64 bits. La última cifra siempre se pone en duda y esta varía de 2 en 2, por ejemplo cosideremos el número 1e16 cuya representacieon binaria es (ver rutinas `bin64_to_float(b)` y `to_bin64()` más abajo)

     0 10000110100 0001110000110111100100110111111000001000000000000000

Si cambiamos el último bit a 1, da en base decimal el número 1.0000000000000002e+16. Note que 1.0000000000000001e+16 tiene la misma representación binaria que 1e16 y 1.0000000000000003e+16 la misma de 1.0000000000000002e+16, pues cambiar el último bit es equivalente a sumar 2 en el último dígito de la representación decimal, esto debido a representación binaria desde donde se llega. 

El valor real asumido por un número de 64-bits con "double-precision" se convierte a base decimal de la siguiente manera:

$$(−1)^s(1 + f)2^{c−1023},$$

más explicitamente, si en base dos $c=(b_{62}b_{61}...b_{52})_2$ y $f=(0.b_{51}b_{50}...b_{0})_2$, entonces,

$$(-1)^{\text{s}}(1.b_{51}b_{50}...b_{0})_2 \times 2^{(b_{62}b_{61}...b_{52})_2-1023},$$

para convertir a notación en base 10 se hace,

$$(-1)^{\text{s}}\left(1 + \sum_{i=1}^{52} b_{52-i} 2^{-i} \right)\times 2^{c-1023}.$$


**Ejemplo**, considere el número binario de 64 bits
     
     0 10000000011 1011100100010000000000000000000000000000000000000000

El primer bit es el signo y los siguientes 11 el exponente, se calcula así:

$$c = 1\times 2^{10} + 0\times2^9 +\cdots+0\times2^2 + 1\times2^1 + 1 \times 2^0 = 1024 + 2 + 1 = 1027$$

el exponencial es entonces $2^{1027−1023} = 2^4$
y la parte fraccional, $f$, de la mantissa es 

$$f = 1\times\left(\frac{1}{2}\right)^1 + 1\times\left(\frac{1}{2}\right)^3+ 1\times\left(\frac{1}{2}\right)^4+ 1\times\left(\frac{1}{2}\right)^5+ 1\times\left(\frac{1}{2}\right)^8+ 1\times\left(\frac{1}{2}\right)^{12} = 0.722900390625$$
el número es 
$$(-1)^0\times2^4(1+0.722900390625)=27.56640625.$$

El número $\pi$ en 64 bits tiene 16 cifras significativas y se representa en binarios como

    3.141592653589793 = 0 10000000000 1001001000011111101101010100010001000010110100011000
    
**Casos especiales para el exponente $c$**:

$c=00000000001_2=1$, es decir: $2^{1 − 1023} = 2^{ − 1022}$ (exponente más pequeño).<br>
$c=11111111110_2=2046$, es decir: $2^{2046-1023}=2^{1023}$ (exponente más grande). <br>
$c=01111111111_2=1023$, es decir: $2^{1023-1023}=2^{0}$ (el exponente es cero). <br>
$c=00000000000_2 =0$: todo el número es cero si $f$ es ceros, sino es número [subnormal](https://es.wikipedia.org/wiki/N%C3%BAmero_subnormal), i.e, el exponente está por debajo del exponente mínimo. <br>
$c=11111111111_2$:  el número es $\infty$ (si $f$ es ceros) o `NaNs` (si $f$ distinto de ceros).

Finalmente es importante mencionar que el tiempo de cómputo con flotantes de 32 bits es aproximadamente tres veces más rápido que con 64, aunque mucho menos preciso y con problemas de redondeo como veremos en la sección de teoría de errores. Los procesadores y sistemas operativos modernos trabajan por defecto con 64 bits, pero las tarjetas graficadoras trabajan por defecto con 32 bits.

# Operaciones bit a bit o bitwise 
Son operaciones sobre números enteros a nivel de sus bits individuales (es decir programación bit a bit) y no pueden ser aplicadas a números tipo float o caracteres,  [los operadores bitwise](https://wiki.python.org/moin/BitwiseOperators) son definidos por 
<<, >>, &, |, ^,  ~, más exactamente, sean $x$ y $y$ números enteros cualesquiera entonces,

1. $x$ << $y$, corre $y$ bits a la izquierda de $x$, equivale a hacer la operación, `x*2**y` (ejemplo 25<<2 = 100). 
2. $x$ >> $y$, corre $y$ bits a la derecha de $x$, equivale a hacer la operación, `x//2**y` (ejemplo 25>>2 = 6).
3. $x$ & $y$,  AND binario (compara bit a bit).
4. $x$ | $y$,  OR binario (compara bit a bit).
5. $x$ ^ $y$,  XOR binario (compara bit a bit).
6. \~$x$,     NOT binario (compara bit a bit), equivale a hacer la operación, `~x = -x-1`.  

Los operadores <<, >>, son una manera rápida de hacer multiplicaciones/divisiones por potencias de dos. En el caso del XOR (OR exclusivo) solo da uno cuando los dos bits son distintos, si son iguales da cero. La operación NOT (~) es conocida como operación complemento e invierte todos los bits incluido el que representa el signo. Cómo ejemplo considere las siguientes operaciones según las tablas de verdad aplicados bit a bit, entre los bits superiores y los inferiores: 
```c
       0011        1011        1010        
    OR 1000    AND 1110    XOR 1001    NOT 1000 
     = 1011      = 1010      = 0011      = 0111
```     
Así como el operador += se puede usar para asignar sumas, se puede hacer lo mismo con los operadores bitwise, por ejemplo, la operación `x>>=1` es equivalente a hacer `x = x>>1`, más concretamente si, 
```python
x=1 entonces x>>=1 da x=0
```
Manipular bits puede resultar en operaciones muy eficientes computacionalmente, a continuación algunos ejemplos:    

````python
#---- operaciones lógicas con bitwise -------------
A = int('00100011',2) # Out: 35
B = int('00101101',2) # Out: 45

print('AND', bin(A & B)) # Out: 0b100001
print('OR' , bin(A | B)) # Out: 0b101111
print('XOR', bin(A ^ B)) # Out: 0b1110
print('NOT', bin(~A),~A) # Out: -0b100100 = -36
````

````python
print('-------- correr bits a la derecha ----------')
print('A:', bin(A))
print(bin(A>>1)) # Out: '0b10001' = 35//2
print(bin(A>>2)) # Out: '0b1000'  = 35//4
print(bin(A>>3)) # Out: '0b100'   = 35//8
print('-------- correr bits a la izquierda --------')
print(bin(A<<1)) # Out: '0b1000110'  = 35*2 
print(bin(A<<2)) # Out: '0b10001100' = 35*4
print(bin(A<<3)) # Out: '0b100011000'= 35*8
````

````python
# min y max de dos valores enteros usando bitwise
# Cuidado: esto redefine (sobrescribe) las funciones nativas min() y max()
# de python para el resto de la sesion; si no quieres perder las nativas,
# usa otro nombre, p.ej. bit_min, bit_max.
min = lambda x,y: y^((x^y)&-(x<y))
max = lambda x,y: x^((x^y)&-(x<y))

min(45,4), max(45,4)
````

````python
# intercambiar valores de manera eficiente entre 2 variables
a = 1; b = 2
def swap(a,b):
    a = a^b
    b = a^b
    a = a^b
    return a,b
swap(a, b)    # note que ahora a = 2 y b = 1
````

````python
# Calcula el Maximo Común Divisor entre dos números,
# osea el número más grande que los divide.
def MCD(a,b):
    while True:
        a%=b; b^=a; a^=b; b^=a #b^=a^=b^=a%=b
        if (b==0): break
    return a

MCD(8,20)
````

### Máscara usando bitwise
Los operadores bitwise se pueden usar para seleccionar elementos de un array en python, por ejemplo se define una máscara como la condición, `mask = (condicion1 & condicion2)`, y luego se inserta como indice del array al que queremos seleccionar, ósea, `array[mask]` solo imprimirá los elementos para los cuales mask es verdadero, en general mask es un arreglo con elementos lógicos, ejemplo: 

````python
# ----------mascara usando bitwise --------------------
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,101,1)
y = 300 - 0.1*np.arange(0,101,1)
mask = (x >= 50) & (x <= 100) # solo seleccione los elementos entre 50 y 100, 
# mask = (x < 40) | (x > 50)  # los parentesis son importantes, quitarlos da error.

print (mask) # imprime array con True of False como elementos.

fig, ax = plt.subplots()
ax.scatter(x[mask], y[mask]) # note que plot comienza en 50 y no en cero.
````

6) Para pasar de representación decimal a binaria o viceversa hay varias formas (no hay que reinventar la rueda), una es usar la librería python [struct](https://docs.python.org/3/library/struct.html). Use las siguientes rutinas para comprobar sus resultados calculados manualmente:

````python
# Convertir binario IEEE 754 de 32 bits a base 10 
import struct
def bin32_to_float(b):
    f = int(b,2)
    x = struct.unpack('f',struct.pack("I",f))[0]
    return x

bin32_to_float('11000100101001000010101000000000')
````

````python
# Convertir binario IEEE 754 de 64 bits a float base 10 
import struct
def bin64_to_float(b):
    f = int(b,2)
    x = struct.unpack('d',struct.pack("<Q",f))[0] # "L"
    return x

# ejemplo con el numero pi
bin64_to_float('0100000000001001001000011111101101010100010001000010110100011000')
````

````python
# Convertir número a representación IEEE 754 de 32 bits: 
import struct
def to_bin32(x):
    return ''.join('{:0>8b}'.format(c) for c in struct.pack('!f', x))

# Convertir número a representación IEEE 754 de 64 bits: 
def to_bin64(x):
    return ''.join('{:0>8b}'.format(c) for c in struct.pack('!d', x))

print("32 bits IEEE 754: -1313.3125 =", to_bin32(-1313.3125))
print("64 bits IEEE 754: -1313.3125 =", to_bin64(-1313.3125))
````

````python
# Ejemplo
# Alternativa: Convertir número flotante (no enteros) a representación IEEE de 64 bits: 
def float_to_bin64(x):
    if x == 0: return "0"*64
    w, signo = (float.hex(x), 0) if x > 0 else (float.hex(x)[1:], 1)
    mantisa  = int(w[4:17], 16) 
    exp      = int(w[18:])
    return "{}{:011b}{:052b}".format(signo, exp + 1023, mantisa)

# ejemplo con el numero pi
print(to_bin64(-3.141592653589793))      # imprimir pi, función anterior
print(float_to_bin64(-3.141592653589793))# imprimir pi, función alternativa
print(float_to_bin64(-1.)) # poner punto para convertir entero en flotante!
````

````python
# Ejemplo
# Note que (2) == (3) y (4) == (5) cuando se pasan a binarios.
print(to_bin64(27.566406249999996)) # (1)
print(to_bin64(27.566406250000004)) # (2)
print(to_bin64(27.566406250000003)) # (3)
print(to_bin64(27.566406250000001)) # (4)
print(to_bin64(27.566406250000000)) # (5)
````

9) *Teoría de conjuntos*: en python un conjunto es una colección de elementos entre corchetes, {}, que no está ordenada ni indexada y cada elemento es único e inmutable (con el comando [set](https://www.programiz.com/python-programming/set), se pueden transformar listas o tuplas a conjuntos, así `set([1,2,3]) = {1,2,3}`, o también es útil para remover elementos repetidos y ordenar listas o arrays: `list(set([1,1,3,2,1])) da [1, 2, 3]`), las operaciones con bitwise pueden ser usadas para hacer operaciones de conjuntos, esto es:

* Operador unión | combina dos conjuntos y elimina los términos repetidos.
* Operador intersección & da el conjunto con los elementos de la intersección.
* Operador de diferencia simétrica ^ da elementos fuera de la intersección en los dos conjuntos.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def ieee(bits, cortes, colores, textos, alto=1, name="Fig"):
    fig, ax = plt.subplots(figsize=(16, .8))

    for i, b in enumerate(bits):
        ax.add_patch(Rectangle((i, 0), 1, alto, fc=colores[i], ec="white"))
        ax.text(i+.5, alto/2, b, ha="center", va="center", fontsize=14)

    y = alto + .25
    for j, (ini, fin) in enumerate(zip(cortes[:-1], cortes[1:])):
        ax.plot([ini, fin], [y, y], color=colores[ini], lw=2)
        ax.text((ini+fin)/2, y+.12, textos[j], ha="center", fontsize=13)

    ax.set(xlim=(-.5, len(bits)+.5), ylim=(-.1, alto+.7))
    ax.axis("off")
    plt.savefig("../figures/" + name + ".png", transparent=True, format="png")


# Float32: −1313.3125
ieee("11000100101001000010101000000000",
     [0, 1, 9, 32],
     ["paleturquoise"] + ["lightgreen"]*8 + ["lightpink"]*23,
     ["Signo", "Exponente (8 bits)", "Mantisa (23 bits)"],1, "Float_example_1")

# Double: 27.56640625
ieee("0100001101000001110000110111100100110111111000001000000000000000",
     [0, 1, 12, 64],
     ["paleturquoise"] + ["lightgreen"]*11 + ["lightpink"]*52,
     ["Signo", " Exponente (11 bits)", "Mantisa (52 bits)"], 1, "Double_example_1")
````
