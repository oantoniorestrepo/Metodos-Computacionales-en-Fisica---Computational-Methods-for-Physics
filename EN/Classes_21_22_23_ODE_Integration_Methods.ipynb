## 1) Basic definitions

### Definition of a differential equation
An equation that has derivatives of any order in its terms is a differential equation; for example, consider the following equations and their solutions:
$\newcommand{\sen}{\,\hbox{sen}\,}$

$$
\begin{array}{l}
1)\quad \dfrac{dy}{dt}= -ky &\iff& y(t)=Ae^{-kt}\\
2)\quad \dfrac{d^2y}{dt^2}+ \omega_0^2y= 0  &\iff&  y(t)=A\cos(\omega_0 t) + B\sin(\omega_0 t),\\
3)\quad \dfrac{d^2y}{dt^2}+ \omega_0 \sin y= A\cos(\alpha t-\pi/3)&\iff& \text{No analytical solution},\\
4)\quad \dfrac{d^3y}{dt^3}+t^2y\dfrac{d^2y}{dt^2} - 3y(y+t)=0&\iff& \text{No analytical solution.}
\end{array}
$$

As we can see, not all of them have an analytical solution, but there are several numerical methods to solve these equations.

### Order and classification of differential equations

The order is determined by the highest derivative in the equation, so equation 1) is first order, 2) and 3) are second order, and the last one is third order. On the other hand,
the classification is given according to whether the equation is linear or non-linear: an ODE is linear if the variable to be integrated can be separated from the operator $\mathcal{L}$ formed by the derivatives, i.e. a differential equation is linear if,

$$\mathcal{L}y(t)=f(t),$$ 

for example, for the first ODE,

$$ \left(\frac{d}{dt} + k\right)y(t)=\mathcal{L} y(t)=0.$$

Thus, equations 1) and 2) are linear but 3) and 4) are not.

### Systems of first-order ODEs

A system of $n$ differential equations is linear (i.e. there are no terms like $y_i^2, 1/y_i$ or $y_iy_j$, for example) if it can be written as,

$$\dfrac{dy_1}{dt}=a_{11}y_1 + a_{12}y_2 + \cdots a_{1n}y_n,$$
$$\dfrac{dy_2}{dt}=a_{21}y_1 + a_{22}y_2 + \cdots a_{2n}y_n,$$
$$\vdots$$
$$\dfrac{dy_n}{dt}=a_{n1}y_1 + a_{n2}y_2 + \cdots a_{nn}y_n.$$

If we define $\mathbf{y}(t)= (y_1(t),y_2(t),..., y_n(t))$, the matrix form and its solution are, 

$$\quad \frac{d\mathbf{y}}{dt}= A\mathbf{y}\iff \mathbf{y}=\exp(At)\mathbf{y}_0$$

where $\mathbf{y}_0=\mathbf{y}(t_0)$ is the initial condition and $A$ is an $n\times n$ matrix; the solution is obtained via the matrix diagonalisation methods. If $A(t)$ varies with time then the solution is, 

$$\mathbf{y}(t)=\exp\left(\int_{t_0}^tA(t')dt'\right)\mathbf{y}_0.$$

In the case of non-linear systems, the coupled equations are written in the form,

$$\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}),$$

note that $\mathbf{f}(t,\mathbf{y})\neq A\mathbf{y}$; these ODE systems have no analytical solution (only in a few cases) and must be solved numerically. Systems such as Edward Lorenz's or the harmonic oscillator for large angles are examples of non-linear systems; in this document we'll show the different numerical methods for solving ODE systems: Euler's method, the Verlet and leapfrog methods, and Runge-Kutta (rk2, rk4 and rk45). Before looking at ODE systems, let's first look at the single-variable case in the following section.

## 2) Euler's method
Consider the differential equation,

$$\frac{dy}{dt}= f(t, y),  \,\,\,   a \le t \le b, \,\,\, y(a) = \alpha.$$

By Taylor expansion of $y(t)$ up to order two, and using the above equation to replace the derivative $y'(t_n)$ with $f (t_n, y(t_n))$, we have, 

$$y(t_{n+1}) = y(t_n) + hf (t_n, y(t_n)) + \frac{h^2}{2} y''(\xi_n),$$

therefore Euler's method is defined as,

$$y_{n+1} \approx y_n + hf(t_n,y_n).$$

The "*local error*" or "*local truncation error*" measures, at each step, the amount by which the approximate solution fails to satisfy the differential equation, and according to the differential equation, it is defined by,

$$\boxed{\tau(h) \overset{\text{def}}{=} \frac{y_{n+1} - y_n}{h} - f(t_n,y_n)=\frac{h}{2}y''(\xi_n).}$$

so this method is first order $O(h)$.

#### Example: Pendulum
Consider a pendulum formed by a mass $m$ tied to a string of length $l$; the string makes an angle $\theta$ with the vertical. If we decompose the force into its radial and tangential components along the trajectory and use Newton's equations, we get

$$
\begin{align}
T-mg\cos\theta =&\,\frac{v^2}{l}\\
-mg \sin\theta =& \,m\frac{d^2s}{dt^2} 
\end{align}
$$

where $s$ is the arc length travelled; if we note that $s=l\theta$, this gives us the non-linear differential equation,

$$
 \frac{d^2\theta}{dt^2} + \omega_0^2\sin\theta = 0,
$$

where $\omega_0 = \sqrt{\frac{g}{l}}$. This equation has an analytical solution if we consider $\theta$ small; in that case $\sin\theta\approx \theta$, which gives the harmonic-oscillator equation (for example, note that the relative error for $\theta=15^\circ$ is ~$1.15\%$ and for $\theta=30^\circ$ it is ~$5\%$, so a value of $\theta<15^\circ$ is small in the sense that it gives an error below one percent; verify this), but for large angles there is no analytical solution. 
Since the equation is second order and Euler's method is for first-order equations, we first write it as two coupled first-order linear equations,

$$
\begin{align}
 \frac{d\theta}{dt} =&\, \omega,\\
 \frac{d\omega}{dt} =&\, - \omega_0^2\sin\theta,
\end{align}
$$

and implementing Euler's method on each equation,

$$
\begin{align}
 \theta_{i+1} =&\,\theta_i + h\omega_i,\\
 \omega_{i+1} =&\,\omega_i - h\omega_0^2\sin\theta_i,
\end{align}
$$

but for better convergence the new $\theta_{i+1}$ is usually used in the second equation, i.e. 

$$\omega_{i+1} =\,\omega_i - h\omega_0^2\sin\theta_{i+1}.$$ 

Now let's look at the numerical solution implemented with Euler's method:

#### Example: Planetary system
Planetary motion is described by Newton's law of universal gravitation,

$$F=-G\frac{mM}{r^2},$$

where $G$ is the universal gravitational constant. The motion of a planet of mass $m$ with coordinates $\mathbf{r}=(x,y)$, around the sun (mass $M$), can be described by the equation,

$$
\mathbf{F} = m\mathbf{a}=m\frac{d^2\mathbf{r}}{dt^2},
$$

whose components are $\mathbf{F}=(f_x,f_y)$, so,

$$
r = \sqrt{x^2+y^2},\\
f_x = F\cos(\theta)=F\frac{x}{r},\\
f_y = F\sin(\theta)=F\frac{y}{r},\\
$$

this gives the system of equations,

$$
\frac{d^2x}{dt^2} = -GM\frac{x}{r^3},\quad\text{and}\quad\frac{d^2y}{dt^2} = -GM\frac{y}{r^3}.
$$

which can be rewritten as the system of first-order linear equations, 

$$
\begin{eqnarray}
\frac{dx}{dt}   &=& v_x            &&    x(t+h)=x(t)+ hv_x(t),\\ 
\frac{dy}{dt}   &=& v_y            &&    y(t+h)=y(t)+ hv_y(t),\\
\frac{dv_x}{dt} &=& -GM\frac{x}{r^3} &\overset{\quad\text{ Euler }\quad}{\Longleftrightarrow} &v_x(t+h)=v_x(t)-hGM\frac{x(t)}{r^3}, \\
\frac{dv_y}{dt} &=& -GM\frac{y}{r^3} && v_y(t+h)=v_y(t)-hGM\frac{y(t)}{r^3},
\end{eqnarray}
$$

finally we define, $y_{0}=x$, $y_{1}=y$, $y_{2}=v_x,$ and $y_{3}=v_y$, which has the vector form $\mathbf{y}'=\mathbf{f}(t,\mathbf{y})$ and is solved with Euler's method.

## 3) Verlet and leapfrog methods for Newton's equations
In 1967 Verlet proposed a second-order method based on Taylor expansion, known as the *Verlet method*. The
Verlet method for integrating Newton's equations is obtained from the expansions,

$$
\begin{align}
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac{a(t)\Delta t^2}{2}+\frac{b(t)\Delta t^3}{6}+O(\Delta t^4)\\
x(t-\Delta t) &=x(t)-v(t)\Delta t+\frac{a(t)\Delta t^2}{2}-\frac{b(t)\Delta t^3}{6}+O(\Delta t^4), 
\end{align}
$$

adding gives,

$$x(t+\Delta t) = 2x (t) - x (t-\Delta t) + a (t) \Delta t^2 + O(\Delta t^4),$$

and the velocity is obtained by central difference,

$$v(t) =\frac {x(t+\Delta t)-x(t-\Delta t)}{ 2 \Delta t },$$

since at the first integration step ($i=0$) $x(-\Delta t)$ is not known, the approximation $x_1=x_0+v_0\,\Delta t+\frac12 a(x_0)\,\Delta t^2$ is used, where $(x_0,v_0)$ are the initial conditions, and for the remaining steps ($i=1,2,\dots$) the two previous formulas are used (by symmetry you can also do $x(-\Delta t)=x_0+v_0\,(-\Delta t)+\frac12 a(x_0)\,\Delta t^2$ and start $i$ from zero, $i=0,1,2,\dots$).
Note that although the local error in position is of order 4, in velocities it is of order 2, so the overall error is of order 2. Also, velocity and position are computed at different times, which isn't good if you want to compute the energy $E=K+V$. 
A better approximation is to use the *leapfrog* method, defined by,

$$
\begin{align}
v(t + \Delta t/2) &= v(t) + a(t)(\Delta t/2)\\
x(t + \Delta t)   &= x(t) + v(t + \Delta t/2)\Delta t\\
v(t + \Delta t)   &= v(t + \Delta t/2) + a(t + \Delta t)(\Delta t/2)
\end{align}
$$

Another way to write the leapfrog method is to use the first equation to replace $v(t + \Delta t/2)$ in the second and third, which gives the more compact form,

$$
\begin{align}
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac {1}{2}a(t)\Delta t^2,\\ 
v(t+\Delta t)&=v(t)+\frac {a(t)+a(t+\Delta t)}{2}\Delta t.
\end{align}
$$

This form is known as *velocity Verlet*. Note these two equations can also be obtained, the first by Taylor-expanding $x(t+\Delta t)$ up to second order and the second for $v(t+\Delta t)$ using the trapezoidal integration method (recall: $v_b-v_a =\int_a^b f(t)\,dt\approx (f(b)+f(a))\Delta t/2$). 

In general, if we consider the force computation, $a(t)=F(x(t))/m$ and $ a(t+\Delta t)=F(x(t+\Delta t))/m$, then

$$
\begin{align}
         a(t)&=F(x(t))/m\\
x(t+\Delta t)&=x(t)+v(t)\Delta t+\frac {1}{2}a(t)\Delta t^2\\ 
a(t+\Delta t)&=F(x(t+\Delta t))/m\\
v(t+\Delta t)&=v(t)+\frac {a(t)+a(t+\Delta t)}{2}\Delta t,
\end{align}
$$

or, changing to index notation,

$$
\boxed{
\begin{align}
  a_i     &= F_i/m\\
  x_{i+1} &= x_i + v_i\, \Delta t + \tfrac{1}{2}\,a_i\, \Delta t^{\,2}, \\
  a_{i+i} &= F_{i+1}/m\\
  v_{i+1} &= v_i + \tfrac{1}{2}(a_i + a_{i+1})\,\Delta t.
\end{align}
}
$$

For $i=0,1,2,\dots,N$ integration steps and $t=i \Delta t$. Verlet algorithms, in all their forms, are time-reversible, i.e. if $t$ is swapped for $-t$, the initial conditions are recovered; this is of great importance for energy computation — integration algorithms that conserve energy are known as *symplectic methods*.

## 4) Second-order Runge-Kutta method (rk2)
Consider the initial value problem,

$$ \frac{dy}{dt}=y'=f(t,y),\quad\hbox{ with }\quad a\leq t\leq b, \quad y(a) = \alpha, $$

which can be solved with the Runge-Kutta-2 algorithm, with truncation error $O(h^2)$ (recall Euler is order $O(h)$). To derive the rk2 method we first integrate the previous equation on the interval $[t,t+h]$, so,

$$y(t+h)=y(t)+\int_t^{t+h}f(t',y(t'))dt'$$

Then we Taylor-expand around the midpoint $t_{h/2}=t+h/2$, 

$$f(t',y(t'))=f(t_{h/2},y(t_{h/2}))+(t'-t_{h/2})\frac{d}{dt}f(t_{h/2},y(t_{h/2}))+O(h^2),$$

The integral of the first term is trivial and gives $hf(t_{h/2},y(t_{h/2}))$, and the integral of the second term is zero, as can be seen (since it's an odd function and the integration interval is symmetric about the midpoint $t_{h/2}$),

$$
\begin{align}
\int_t^{t+h} (t'-t_{h/2})\,dt'
  =&\,\frac{1}{2}(t'-t_{h/2})^2\Big|_t^{t+h} \\
  =&\,\frac{1}{2}\left(t+h-t-\frac{h}{2}\right)^2-\frac{1}{2}\left(t-t-\frac{h}{2}\right)^2\\
  =&\,0.
\end{align}
$$

So we obtain

$$y(t+h)=y(t)+hf(t_{h/2},y(t_{h/2}))$$

to know $y(t_{h/2})$ we Taylor-expand around $t$ to first order, which gives 

$$y(t_{h/2})\approx y(t)+(t_{h/2}-t)\frac{d}{dt}y(t)=y(t)+\frac{h}{2}f(t,y(t)).$$

where we used $y'=f(t,y(t))$ and $(t_{h/2}-t)=h/2$; finally, if we define the constants $k_1$ and $k_2$ we get the rk2 algorithm,

\begin{align}
k_1   =&\, hf(t,y)\\
k_2   =&\, hf\left(t+\frac{h}{2},y+\frac{k_1}{2}\right)\\
y(t+h)=&\, y(t)+k_2.
\end{align}

This particular form is known as the midpoint method. 
Implementation:


````python
def rk2( f, y, t, h ):

    k1 = h*f(t, y)
    k2 = h*f(t + 0.5*h, y + 0.5*k1)
    y = y + k2

    return y
````

### Systems of first-order differential equations
Suppose we have a system of $N$ particles interacting with each other; each particle follows Newton's laws, so from the second law we get a system of $3N$ coupled equations — this is what in physics is known as the $N$-body problem, and unfortunately it only has an analytical solution for $N\le2$, so numerical methods must be implemented to solve this problem if $N\ge3$.
Runge-Kutta can be used to solve systems of equations; for this the equations must first be put in first-order linear vector form,  

$$
\begin{pmatrix}
  \frac{d}{dt}y_1(t)\\
  \frac{d}{dt}y_2(t)\\
  \vdots\\
  \frac{d}{dt}y_n(t)\\
\end{pmatrix}
=
\begin{pmatrix}
  f_1(t,y_1,y_2,\cdots,y_n)\\[7pt]
  f_2(t,y_1,y_2,\cdots,y_n)\\
  \vdots\\
  f_n(t,y_1,y_2,\cdots,y_n)\\
\end{pmatrix}
\quad\Longrightarrow\quad
\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}),
$$

with $a\leq t\leq b,$ and $\mathbf{y}(a) = \boldsymbol{\alpha}$, which is the desired form to solve numerically with Runge-Kutta.


### Higher-order differential equations
The order of a differential equation is given by its highest-order derivative; so, for example, Newton's equations are second order. In general a linear equation of order $n$ has the form,

$$a_n(t)\frac{d^ {n}}{dt^n}y(t)+a_{n-1}(t)\frac{d^{n-1}}{dt^{n-1}}y(t)+\cdots+a_1(t)\frac{d}{dt}y(t)+a_0(t)y(t)=f(t),$$

where the $a_i$ are functions of $t$; this equation can be transformed into a linear system of $n$ first-order equations if we carry out the following transformations: define $y=y_0$ and a new variable $y_i$ for each derivative — this transforms the order-$n$ equation into a system of $n$ coupled first-order equations. Suppose for simplicity that $a_n(t)=1$; this gives the system,

$$
\begin{align}
  \frac{d}{dt}y_0(t)=&\, y_1,\\
  \frac{d}{dt}y_1(t)=&\, y_2,\\
  \vdots\\
  \frac{d}{dt}y_{n-1}(t)=&\, f(t)-\left[a_{n-1}(t)y_{n-1} +\cdots +a_1(t)y_1 +a_0(t)y_0\right],   
\end{align}
$$

where the $n$ variables, $y_0,\cdots,y_{n-1}$, are numbered from 0 to $n-1$ to be friendlier with python's array notation.

**Example** consider the order-3 equation,

$$t^2\frac{d^ {3}}{dt^3}y(t)+\,\hbox{sen}(t+2)\frac{d^{2}}{dt^{2}}y(t)+e^{2t}\frac{d}{dt}y(t)+\cos(t)=0,$$

if we divide the equation by $t^2$ we get $a_2(t)=t^{-2}\,\hbox{sen}(t+2)$, $a_1(t)=t^{-2}e^{2t}$, $a_0(t)=0$, and $f(t)=-t^{-2}\cos(t)$, which gives the system

$$
\begin{align}
  \frac{d}{dt}y_0(t)  =&\, y_1,\\
  \frac{d}{dt}y_1(t)  =&\, y_2=\,\frac{d}{dt}\left(\frac{d}{dt}y_0(t)\right),\\
  \frac{d}{dt}y_{2}(t)=&\, f(t) -a_{2}(t)y_{2} -a_1(t)y_1 .   
\end{align}
$$

Substituting into vector form gives,

$$
\begin{pmatrix}
  \frac{d}{dt}y_0(t)\\
  \frac{d}{dt}y_1(t)\\
  \frac{d}{dt}y_2(t)\\
\end{pmatrix}
=
\begin{pmatrix}
\, y_1\\[5pt]
\, y_2\\
\, -t^{-2}(\,\hbox{sen}(t+2)y_{2} +e^{2t}y_1 +\cos(t))\\
\end{pmatrix}
$$

in which we recognise the system $\mathbf{y}' =\mathbf{f}(t,\mathbf{y})$, which can be solved with Runge-Kutta. Note that for the case of a system of $N$ Newton's equations, each equation must first be put as a system of two first-order equations, i.e. one ODE for the velocity and another for the position, for each dimension, and since space is three-dimensional, the $N$-body problem gives a system of $6N$ first-order equations to solve.

## 5) Fourth-order Runge-Kutta method (rk4)
Runge-Kutta-2 is almost never used and is only shown for teaching purposes; higher-order methods are used instead. Runge-Kutta-4 (rk4) is one of the most popular methods and has a numerical truncation error of order $O(h^4)$; the derivation is similar to rk2 but much longer and is omitted ([see the derivation on wikipedia](https://en.wikipedia.org/wiki/Runge%E2%80%93Kutta_methods#Derivation_of_the_Runge.E2.80.93Kutta_fourth-order_method)). 
Consider the system of $N$ coupled equations (in bold to remind us of the vector character, see example below),

$$\frac{d\mathbf{y}}{dt}=\mathbf{y}' =\mathbf{f}(t,\mathbf{y}) ,\\ \\ \\ a\leq t\leq b, \\ \\ \\ \\ \mathbf{y}(a) = \boldsymbol{\alpha},$$

where $\mathbf{y}(t)=\{y_1(t),y_2(t),\cdots,y_N(t)\}$ are the variables to integrate and $\mathbf{f}(t,\mathbf{y})=\{f_1,f_2,\cdots,f_N\}$.
The rk4 iterative method uses the following expressions to solve this system

$$ 
\begin{align}
 \mathbf{k}_1 =&\, h\mathbf{f}\,(t,\mathbf{y})\\
 \mathbf{k}_2 =&\, h\mathbf{f}\left( t + \frac{h}{2},\mathbf{y} + \frac{\mathbf{k}_1}{2}\right)\\
 \mathbf{k}_3 =&\, h\mathbf{f}\left( t + \frac{h}{2},\mathbf{y} + \frac{\mathbf{k}_2}{2}\right)\\
 \mathbf{k}_4 =&\, h\mathbf{f}\left( t + h,\mathbf{y} + \mathbf{k}_3\right)\\
 \mathbf{y}(t+h) =\,& \mathbf{y}(t) + \frac{1}{6}( \mathbf{k}_1 + 2\mathbf{k}_2 + 2\mathbf{k}_3 + \mathbf{k}_4  )
 \end{align}
$$

Python implementation:


````python
def rk4(f,y,a,b,n):
    '''------------------ Runge-kutta 4  ------------------------
    solves a system with several differential equations (ode)
    '''
    t = a
    d  = len(y)      # Get number of equations.
    k1 = k2 = k3 = k4 = zeros(d)
    
    tt = zeros(n)    # Time array.
    Y  = zeros((d,n))# Coords array.
    h = (b - t)/n
    ti = 0
    while ti<n:

        Y[:, ti] = y # Store coords at time t_i as a column
        tt[ti] = t   # Store the time.

        # *** Runge-Kutta 4 method ***   
        k1 = h*f(t, y)
        k2 = h*f(t + h/2., y + k1/2.)
        k3 = h*f(t + h/2., y + k2/2.)
        k4 = h*f(t + h, y + k3)
        
        y = y + (k1 + 2.*(k2 + k3) + k4)/6.
    
        t = t + h   # Increment time by h.
        ti = ti + 1 # Increment counter.  
        
    return Y,tt     # Output: arrays with coords and time for plotting.
````

### Stability of differential equations
This refers to how much the solution can change if a small change is made to the initial conditions: "*in chaotic systems, small changes can give completely different solutions*".
To numerically check how sensitive an equation (or system of equations) is, we check whether a
small change in the initial conditions grows exponentially, i.e. if we take
$x_0$ and $y_0=x_0+\epsilon$, then $|x(t)−y(t)|$ is defined and we check whether this quantity grows exponentially; if this is the case the system is chaotic (more formally, for $\epsilon\to0$ and $t\to\infty$ the exponent of this dependence, if it exists, is known as the Lyapunov exponent).

### Attractor
An [attractor](https://en.wikipedia.org/wiki/Attractor) is a set of numerical values towards which a system of equations tends to evolve, over a wide range of the system's initial conditions. For example, dynamical systems in physics that are dissipative (due to internal friction, thermodynamic losses, mass loss, etc.) show attractor behaviour.

**Exercise** In 1963 Edward Lorenz developed a mathematical model for computing atmospheric instability; this model is known as [the Lorenz attractor](https://en.wikipedia.org/wiki/Lorenz_system#Python_simulation) and is defined by the following equations,

$$
\begin{align}
\frac{\mathrm{d}x}{\mathrm{d}t} &= \sigma (y - x), \\
\frac{\mathrm{d}y}{\mathrm{d}t} &= x (\rho - z) - y, \\
\frac{\mathrm{d}z}{\mathrm{d}t} &= x y - \beta z.
\end{align}
$$
which has chaotic behaviour for certain values of $\sigma, \rho$ and $\beta$.
The following routine uses the scipy library's function `scipy.integrate.odeint`, implemented for integrating differential equations, and then we compare the result against rk4: 


````python
# 1) use scipy to integrate
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint      # scipy integration method
from mpl_toolkits.mplot3d import Axes3D # make a 3D plot

rho = 28.0
sigma = 10.0
beta = 8.0 / 3.0

def f(state, t):
  x, y, z = state             # create a tuple of the form (x,y,z)       
  return sigma * (y - x), x * (rho - z) - y, x * y - beta * z  

state0 = [1.0, 1.0, 1.0]      # initial condition
t = np.arange(0.0, 40.0, 0.01)

states = odeint(f, state0, t) # Integrate the ode system, dy/dt = f(y, t0, ...)

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(states[:,0], states[:,1], states[:,2])
plt.show()
````

````python
# 2) use rk4 to integrate
# Initialisation
a, b = 0., 40.         # Interval [a,b] 
n = 10000              # max number of integration steps
y0 = array([1.,1.,1.]) # initial conditions 

rho = 28.0; sigma = 10.0; beta = 8.0/3.0
fR = zeros(3)          # initialise array

# function f(t,y) = sigma * (y - x), x * (rho - z) - y, x * y - beta * z
def f(t, y):
    fR[0] = sigma * (y[1] - y[0]) 
    fR[1] = y[0] * (rho - y[2]) - y[1]
    fR[2] = y[0] * y[1] - beta * y[2]
    return fR
    
Y, time = rk4(f,y0, a,b, n)  # integrate
    
fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot(Y[0], Y[1], Y[2])
ax.set_xlabel('X(t)')
ax.set_ylabel('Y(t)')
ax.set_zlabel('Z(t)')

plt.show()
````

Note that for $\rho = 28.0, \sigma = 10.0$ and $\beta = 8.0/3.0$ the Lorenz attractor produces completely different solutions with small variations in the initial conditions $(x_0,y_0,z_0)$ (for example by adding $10^{-5}$ to $x_0$; to see this plot both cases and compare). Another interesting case is when $\rho<1$, since all trajectories converge to the origin $(0,0,0)$ (again, plot two trajectories with different initial conditions to see this).
The Lorenz attractor is a classic example of chaotic systems.

### Runge-Kutta-Fehlberg Method (rk45)
This is another popular method that controls the error at each step, and therefore has a variable step size, $h_i$. It's called rk45 because it combines an $O(h^{4})$ Runge-Kutta approximation together with another of order $O(h^{5})$ to compute the error, i.e. we know the absolute error is given by $\text{err} =|y_{i+1} - y|$, where $y$ is the exact solution at $t_i +h$; since $y$ is not known, it is approximated by the value $y\approx z_{i+1}$ of order  $O(h^{5})$. That is, the Runge-Kutta approximation to $O(h^{4})$ is,



$$ 
 \mathbf{y}_{i+1} = \mathbf{y}_{i}  + \frac{25}{216}\mathbf{k}_1 + \frac{1408}{2565}\mathbf{k}_3 + \frac{2197}{4104}\mathbf{k}_4 - \frac{1}{5}\mathbf{k}_5,
$$

 but a better Runge-Kutta approximation of order $O(h^{5})$ is,
 
$$
\mathbf{z}_{i+1} = \mathbf{y}_{i}  + \frac{16}{135}\mathbf{k}_1 + \frac{6656}{12825}\mathbf{k}_3 + \frac{28561}{56430}\mathbf{k}_4 - \frac{9}{50}\mathbf{k}_5 + \frac{2}{55}\mathbf{k}_6,
$$
 
where,

$$
\begin{eqnarray}
\mathbf{k}_1 &=& h\mathbf{f}\left(t_i, \mathbf{y}_i\right), \\
\mathbf{k}_2 &=& h\mathbf{f}\left(t_i + \frac{h}{4}, \mathbf{y}_i + \frac{\mathbf{k}_1}{4}\right),\\
\mathbf{k}_3 &=& h\mathbf{f}\left(t_i + \frac{3}{8}h, \mathbf{y}_i + \frac{3}{32}\mathbf{k}_1 + \frac{9}{32}\mathbf{k}_2\right),\\
\mathbf{k}_4 &=& h\mathbf{f}\left(t_i + \frac{12}{13}h, \mathbf{y}_i + \frac{1932}{2197}\mathbf{k}_1 - \frac{7200}{2197}\mathbf{k}_2 + \frac{7296}{2197}\mathbf{k}_3\right),\\
\mathbf{k}_5 &=& h\mathbf{f}\left(t_i + h, \mathbf{y}_i + \frac{439}{216}\mathbf{k}_1 - 8\mathbf{k}_2 + \frac{3680}{513}\mathbf{k}_3 - \frac{845}{4104}\mathbf{k}_4\right),\\
\mathbf{k}_6 &=& h\mathbf{f}\left(t_i + \frac{1}{2}h, \mathbf{y}_i - \frac{8}{27}\mathbf{k}_1 + 2\mathbf{k}_2 - \frac{3544}{2565}\mathbf{k}_3 + \frac{1859}{4104}\mathbf{k}_4 - \frac{11}{40}\mathbf{k}_5\right),
\end{eqnarray}
$$

the most optimal step is given by $h\rightarrow sh$,

$$
s=\left(\frac{\epsilon h}{2\|\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\|}\right)^{\frac{1}{4}}\approx 
0.84\left(\frac{\epsilon h}{\text{err}}\right)^{\frac{1}{4}},
$$

where the error between the two approximations is given by, $\text{err} =\|\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\|$, but for computational efficiency it's better to compute it as,

$$
\begin{eqnarray}
\text{err} 
&=& \text{max}\left(\mathbf{z}_{i+1} -\mathbf{y}_{i+1}\right),\\
&=& \text{max}\left(\frac{\mathbf{k}_1}{360} - \frac{128}{4275}\mathbf{k}_3 - \frac{2197}{75240}\mathbf{k}_4 + \frac{\mathbf{k}_5}{50} + \frac{2}{55}\mathbf{k}_6\right),
\end{eqnarray}
$$        

where $\text{max}()$ returns the largest vector component in absolute value. Note that to determine the most optimal $h$ at each step, two different approximations of the solution are used, $\mathbf{y}_{i+1}$ and $\mathbf{z}_{i+1}$, and are compared; if the two approximations are similar according to a given tolerance $\epsilon$, $\mathbf{y}_{i+1}$ is accepted and $h$ is increased or reduced by multiplying by $s$; otherwise, $h$ is likewise scaled to $sh$ and $\mathbf{y}_{i+1}$ is recomputed. Note that if `err` is small, $s$ will be large and $h$ will increase, but if `err` is large, $s$ will be small and $h$ will decrease.
For stability and efficiency the values of $s$ and $h$ are restricted to the intervals $0.1 \leq s \leq 4.0$ and  $h_{min} \leq h \leq h_{max}$.
Although rk45 computes the function $f(t,y)$ 6 times, it is in general faster and more precise than rk4 (for more details see Burden's book). The implementation of rk45 with $\epsilon = 10^{-5}$ is:


````python
def rk45(f, y,t,b, hmin = 0.001, hmax = 2., Tol = 1e-5):     
    '''-------------------- Runge-Kutta-Fehlberg (rk45) ------------------------
    Runge-Kutta-Fehlberg method, integrates the vector system, 
                     y'(y) = f(t,y(t)) 
    f: vector function, y: initial condition, (t,b): integration interval, 
    (hmin, hmax): min and max values of h, Tol: error tolerance. 
    '''

    d  = len(y)         # size of the system of equations 
    k1 = k2 = k3 = k4 = k5 = k6 = np.zeros(d)
    
    h = hmax            # initialise h to the max value 
    tt = t              # 1D time array
    Y  = np.array([y])  # 2D array with coords      

    while t < b:
        # *** Runge-Kutta-Fehlberg method (rk45) ***   
        k1 = h*f(t, y)                
        k2 = h*f(t + h/4, y + k1/4)
        k3 = h*f(t + 3/8*h, y + 3/32*k1 + 9/32*k2)
        k4 = h*f(t + 12/13*h, y + 1932/2197*k1 - 7200/2197*k2 + 7296/2197*k3)
        k5 = h*f(t + h, y + 439/216*k1 - 8*k2 + 3680/513*k3 - 845/4104*k4)
        k6 = h*f(t + 1/2*h, y - 8/27*k1 + 2*k2 - 3544/2565*k3 + 1859/4104*k4 - 11/40*k5)
        # Compute the error |z(i+1) − y(i+1)|
        err = (np.abs(k1/360 - 128/4275*k3 - 2197/75240*k4 + k5/50 + 2/55*k6)).max()
        # Accept or reject integration step i+1 
        if (err/h < Tol): 
  
           y = y + 25/216*k1 + 1408/2565*k3 + 2197/4104*k4 - k5/5
           t = t + h               # increment time by h

           Y  = np.append(Y, np.array([y]), axis=0)# store coords at time t_i
           tt = np.append(tt,t)                    # store time in the array
           
        # compute the value of h for the next step, check that h isn't too 
        # large or too small.  
        s = 0.84*(Tol*h/err)**0.25 # scalar to multiply h--> s*h
        
        if (s <= 0.1): h = 0.1*h   # (large err) 0.1 is the min allowed value of s
        elif (s >= 4): h = 4.*h    # 4 is the max allowed value of s
        else:          h = s*h     # scale h to s*h, ideal step ( 0.1 < s < 4)
        
        if   (h > hmax): h = hmax  # don't allow h larger than hmax
        elif (h < hmin): 
            print('(Minimum h exceeded, procedure completed unsuccessfully.')
            break
            
        if (t+h > b): h = b - t    # ensure that on the last step t = b  
                    
    return Y.T, tt # output: arrays with coords and time for plotting.
````
