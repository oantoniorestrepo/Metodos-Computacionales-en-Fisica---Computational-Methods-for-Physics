# Binary representation
Numbers and characters are represented in computing with zeros and ones. A zero or a one is known as a "bit"; these in turn are grouped into packets of 8 bits, which make up a byte.
Memory is measured in bytes:

* kilobyte, 1KB
* megabyte, 1MB
* gigabyte, 1GB

Because base 2 is used instead of base 10, a KB is 1024 bytes (not bits), so 1MB is 1024*1024 bytes and 1GB is 1024$^3$ bytes.

Numeric representation depends on the number of characters or symbols used to represent a number: in decimal base 10 characters are used, in base 2 only 2 are used, and in hexadecimal base 16 symbols are used. The following table compares the three bases,
<a id='Conversions'></a>

Decimal|   Binary|   Hexadecimal
-------|----------|--------------
   0   |   0000   |   0          
   1   |   0001   |   1          
   2   |   0010   |   2          
   3   |   0011   |   3          
   4   |   0100   |   4          
   5   |   0101   |   5          
   6   |   0110   |   6          
   7   |   0111   |   7          
   8   |   1000   |   8          
   9   |   1001   |   9          
  10   |   1010   |   A          
  11   |   1011   |   B          
  12   |   1100   |   C          
  13   |   1101   |   D          
  14   |   1110   |   E          
  15   |   1111   |   F  
  
 Note that with 4 bits you can represent at most 16 numbers (0 through 15); in general, with $N$ bits there are $2^N$ combinations and the largest integer is $2^N-1$. Using 8 bits gives a maximum of $2^8=256$ different combinations, which are used to represent the 256 characters of the [ASCII](https://en.wikipedia.org/wiki/ASCII) code, based on the Latin alphabet; for example the character `a` is represented by 01100001 and `A` by 01000001 (uppercase and lowercase letters differ by only one bit, which makes conversion easier). In the ASCII code the digits 0 through 9 are also represented with 8 bits (to find their representation, from the table above prefix the value 0011 to each of the binaries from 0 to 9; for example zero has ASCII code 00110000). So, for example, a file that only contains the word "hello" is 5 bytes, or 40 bits, in size.


### Binary representation of integers
To convert from base 10 to binary you divide by 2 until you reach 1, example
```c
       445/2      1 
       222/2      0 
       111/2      1
        55/2      1
        27/2      1
        13/2      1
         6/2      0
         3/2      1
         1/2      1
```  
So the binary is 110111101. Let's verify that this binary really is 445:

$$1\times2^8 + 1\times2^7 + 0\times2^6 + 1\times2^5 + 1\times2^4 + 1\times2^3 + 1\times2^2 + 0\times2^1 +1\times2^0=445.$$

An integer in python3 is of type `int` and is implemented at a low level via C's `long` type.
Integers in python3 have arbitrary precision (an integer as large as you like), as can be seen in ipython using the following routine
```python
    long=1
    for n in range(1000):
        long=long*2
        print(n, long) 
```       
in python 2.7 the same loop also gives a 302-digit number (with a trailing `L`, see below), since python 2.7 actually has two distinct integer types: `int`, which maps directly to C's `long`, bounded to 32 (or 64) bits, and `long`, which is genuinely of arbitrary precision and is the type python automatically promotes an `int` to on overflow. The 32-bit C `int` type gives a range $-2^{31}< x < 2^{31} – 1$, (or $-2147483648$ to $2147483647$, i.e. with 10 significant figures); note the exponent is 31 because one bit is reserved for the sign. With 64 bits, the range is $-9223372036854775808 < x< 9223372036854775807$, i.e. with 19 significant figures. In python 3.x this distinction disappears: only `int` exists, and it is always of arbitrary precision (in python 2.7, numbers of type `long` get an `L` appended at the end to distinguish them from `int`; this notation is now obsolete in python 3.x).

### Binary representation of floating-point numbers
Numbers with a decimal point are known as floating-point numbers, and on modern computers they are represented with 32 bits (type float) or 64 bits (type double). (NOTE: the word *float* has 2 meanings depending on context: numbers with a decimal point, and also the 32-bit representation).

The simplest way to convert a decimal-point number to binary is by separating the integer and fractional parts.
For the integer part you divide by 2 and the remainder gives the bits; for the fractional part you multiply by 2, split the result again into integer and fractional parts — the integer part will be 0 or 1 and gives the bits — and the fractional part is divided again until it reaches 1.0.

Example: the number -1313.3125 is converted to binary as follows:
```c
        integer part                     fractional part
     Division     Remainder        Fraction x 2          Integer
                           
     1313/2       1                0.3125 × 2 = 0.625    0 
      656/2       0                0.625  × 2 = 1.25     1 
      328/2       0                0.25   × 2 = 0.5      0 
      164/2       0                0.5    × 2 = 1.0      1
       82/2       0             
       41/2       1             
       20/2       0             
       10/2       0             
        5/2       1             
        2/2       0             
        1/2       1             
```
The result for the integer part is $10100100001_{2}$ and the fractional part is $0101_{2}$. Adding the two parts, the result is $-10100100001.0101_{2}$. To convert from base 2 to base 10 negative powers of 2 are used,

$$0\times2^{-1}+ 1\times2^{-2} + 0\times2^{-3} + 1\times2^{-4} =0.3125$$

### Truncation
Real numbers cannot always be represented on a computer, because many of them require an infinite representation in base 2 (meaning infinite memory to store the number); this is the case, for example, for the numbers $\pi$, $e$, 1/10 or 1/3, since,
```c
    pi   = 3.1416... = 11.001001000011111101101010100010001000010110...
    e    = 2.7182... = 10.101101111110000101010001011000101000101011...
    1/10 = 0.1       =  0.000110011001100110011001100110011001100110...
    1/3  = 0.3333... =  0.01010101010101...
```    
they go on forever in base 2. This means only a fixed number of bits is kept to represent the number, which is known as truncation error, since the truncated number is an approximation of the real number. Note this is not exclusive to base 2: in base 10 the number 1/10 has an exact representation, 0.1, but 1/3 does not have a finite representation, so it can be represented as 0.33 or as 0.333333333, which is better but not exactly 1/3. This affects arithmetic operations, so 0.1\*10 is not exactly 1 but an approximation, similarly in base 10, since 3\*0.33333333 = 0.99999999 and not 1.

### Exponential notation and normalisation
Exponential notation can also be used in base $2$, simply by shifting the point as in decimal notation; for example, the previous number $-1313.3125_{10}$ can be written as,

$$-10100100001.0101_{2}=-1.01001000010101_{2}\times2^{10},$$

specifically, if the point is shifted up to the first 1, one bit to the left, this is known as *normalisation*.

````python
# Convert a base-10 float to binary ()
# Note: for integers it's better to use python's bin() function.

def ftob(x, Tol=20):   # Tol: max number of digits after the point
    
    sig = "-" if x<0 else "" # sign of the number
    n = int(x)         # take integer part of x
    f = abs(x - n)     # take fractional part of x
    
    # *** Compute integer part (divide by 2) 
    L = ""             # build string for the integer part
    while abs(n)>0:
        L = str(n%2)+L # remainder of dividing by 2    
        n = int(n/2)   # integer part of dividing by 2
            
    # *** Compute fractional part (multiply by 2)
    S = ""             # build string for the fractional part
    dot = "." if f != 0 else ""
    while f != 0:
        D = f*2        # multiply fractional part by 2 
        i = int(D)     # take integer part (it is 0 or 1)
        S = S + str(i) # store the bit in the string
        f = D - i      # remove the integer part
        if len(S) > Tol: break # truncate if "max number of digits" > Tol
    
    return sig+L+dot+S # return string of ones and zeros

print(ftob(-1313))
print(ftob(-1313.3125))    
print(ftob(513.01,100))
# ftob(3.1415926)
````

````python
# Convert a "binary" (must be a string) to base 10
def btof(x):
    if isinstance(x,str):# check whether x is a string
       p = x.split('.')  # split integer and fractional parts
       n = list(p[0])    # build list from the binary integer part
       n = n[::-1]       # reverse the list order
       
       # build the base-10 integer part d and fractional part s:
       d = sum([ int(n[i])*2**i for i in range(len(n))])
       num = d
    
       # build the base-10 integer part d and fractional part s:
       if len(p)>1:      # check whether there is a fractional part 
          f = list(p[1]) # build list from the binary fractional part
          s = sum([ int(f[i])*2**(-i-1) for i in range(len(f))]) 
          num = d+s      # add the fractional part
        
       return num        # return the base-10 float
    else:
       print("The binary must be a string, e.g., '10.11'.") 

print(btof('10100100001.0101')) # 1313.3125
print(btof('1000000001'))       # 513
print(btof('1000000001.000000101000111101011'))
print(btof('1000000001.000000101000111101011100001010001111010111'))
print(btof('11.001001000011111101101010100010001000010110'))
btof(ftob(3.1415926)) # or also like this
````

## The IEEE 754 convention
The binary representation above is not the most suitable for the computer, which only sees zeros and ones and not the minus sign or the point between the integer and fractional part. Modern computers use the IEEE 754 convention, in which a number is represented with 16, 32 or 64 bits; let's look at the 32- and 64-bit cases.

**32-bit representation**: here the first bit $b_{31}$ is the sign, the next 8 bits $\{b_{30}b_{29} \dots b_{23}\}$ are the exponent, and the remaining 23 bits are the *base* or *mantissa* written in normalised form (i.e. the integer part is always 1, plus the *fractional part* $f$, represented by the remaining bits); that is:

$$(−1)^s(1 + f)2^{c−127},$$

or more explicitly, if the base-2 exponent is $c=(b_{30}b_{29} \dots b_{23})_2$ and the mantissa is $f=(b_{22}b_{21} \dots b_0)_2$, then,

$$(-1)^{b_{31}} \times (1.b_{22}b_{21} \dots b_0)_2 \times 2^{(b_{30}b_{29} \dots b_{23})_2 - 127},$$

converting to base 10 is simply,

$$(-1)^{b_{31}}\times \left(1 + \sum_{i=1}^{23} b_{23-i} 2^{-i} \right)\times 2^{(c-127)}.$$

Note the mantissa is normalised (hence the "$1.$" followed by the 23 bits representing the fraction $f$); also note the exponent given by $c$ is a positive number between 0 and 255 in base 10, so to guarantee positive and negative exponents 127 is subtracted from it. In memory the bits are stored as follows,

    * Sign: 1 bit (the first)
    * Exponent: 8 bits (the next 8)
    * Mantissa: 24 bits (23 explicitly stored)
  

<img src="../figures/Float_example_2.png" alt="Drawing" style="width: 1000px;"/>

The 32-bit representation gives about 7 significant figures in base 10 (since $2^{24}-1= 16777215$ is the largest mantissa that can be stored — and what is the largest exponent that can be stored?).

**Example**: in the example above we had $-1313.3125_{10}=-10100100001.0101_2$; let's express the result in the more practical IEEE 754 notation:

1) We normalise as explained above: $10100100001.0101 = 1.01001000010101 \times 2^{10}$

2) the fractional part of the mantissa is $01001000010101000000000$, (note the remaining zeros are padded on the right) and the exponent is $c = 10 + 127 = 137 = 10001001$, 

3) The first bit must be 1 since the number is negative. 

Finally $-1313.3125$ is 1 10001001 01001000010101000000000.

**64-bit representation**: in the IEEE 754 standard a "double" is specified by a 64-bit binary as follows (each slot occupies one bit $b_i$ at position $i$):

    * Sign: 1 bit (the first)
    * Exponent: 11 bits (the next 11)
    * Mantissa: 53 bits (52 explicitly stored)

<img src="../figures/Double_example_2.png" alt="Drawing" style="width: 1000px;"/>

This gives a base-10 number with 17 significant figures of precision, or equivalently 16 digits after the point in scientific notation (how do we know this? simply by looking at $2^{53}$ in base 10). In python floats are 64-bit by default. The last digit is always uncertain and it changes in steps of 2; for example consider the number 1e16, whose binary representation is (see the routines `bin64_to_float(b)` and `to_bin64()` below)

     0 10000110100 0001110000110111100100110111111000001000000000000000

If we change the last bit to 1, in decimal this gives the number 1.0000000000000002e+16. Note that 1.0000000000000001e+16 has the same binary representation as 1e16, and 1.0000000000000003e+16 has the same as 1.0000000000000002e+16, since changing the last bit is equivalent to adding 2 to the last digit of the decimal representation — a consequence of the binary representation it comes from. 

The actual value taken by a 64-bit "double-precision" number is converted to base 10 as follows:

$$(−1)^s(1 + f)2^{c−1023},$$

more explicitly, if in base 2 $c=(b_{62}b_{61}...b_{52})_2$ and $f=(0.b_{51}b_{50}...b_{0})_2$, then,

$$(-1)^{\text{s}}(1.b_{51}b_{50}...b_{0})_2 \times 2^{(b_{62}b_{61}...b_{52})_2-1023},$$

to convert to base-10 notation this is done,

$$(-1)^{\text{s}}\left(1 + \sum_{i=1}^{52} b_{52-i} 2^{-i} \right)\times 2^{c-1023}.$$


**Example**: consider the 64-bit binary number
     
     0 10000000011 1011100100010000000000000000000000000000000000000000

The first bit is the sign and the next 11 are the exponent, computed as follows:

$$c = 1\times 2^{10} + 0\times2^9 +\cdots+0\times2^2 + 1\times2^1 + 1 \times 2^0 = 1024 + 2 + 1 = 1027$$

the exponential part is then $2^{1027−1023} = 2^4$
and the fractional part, $f$, of the mantissa is 

$$f = 1\times\left(\frac{1}{2}\right)^1 + 1\times\left(\frac{1}{2}\right)^3+ 1\times\left(\frac{1}{2}\right)^4+ 1\times\left(\frac{1}{2}\right)^5+ 1\times\left(\frac{1}{2}\right)^8+ 1\times\left(\frac{1}{2}\right)^{12} = 0.722900390625$$
the number is 
$$(-1)^0\times2^4(1+0.722900390625)=27.56640625.$$

The number $\pi$ in 64 bits has 16 significant figures and is represented in binary as

    3.141592653589793 = 0 10000000000 1001001000011111101101010100010001000010110100011000
    
**Special cases for the exponent $c$**:

$c=00000000001_2=1$, i.e.: $2^{1 − 1023} = 2^{ − 1022}$ (smallest exponent).<br>
$c=11111111110_2=2046$, i.e.: $2^{2046-1023}=2^{1023}$ (largest exponent). <br>
$c=01111111111_2=1023$, i.e.: $2^{1023-1023}=2^{0}$ (the exponent is zero). <br>
$c=00000000000_2 =0$: the whole number is zero if $f$ is all zeros, otherwise it is a [subnormal](https://en.wikipedia.org/wiki/Subnormal_number) number, i.e. the exponent is below the minimum exponent. <br>
$c=11111111111_2$: the number is $\infty$ (if $f$ is all zeros) or `NaN` (if $f$ is non-zero).

Finally it's worth mentioning that computation time with 32-bit floats is roughly three times faster than with 64-bit floats, though much less precise and prone to rounding issues, as we'll see in the section on error theory. Modern processors and operating systems work with 64 bits by default, but graphics cards work with 32 bits by default.


# Bitwise operations
These are operations on integers at the level of their individual bits (i.e. bit-by-bit programming), and they cannot be applied to floats or characters. [Bitwise operators](https://wiki.python.org/moin/BitwiseOperators) are defined by
<<, >>, &, |, ^,  ~; more precisely, let $x$ and $y$ be any integers, then,

1. $x$ << $y$, shifts $x$ by $y$ bits to the left; equivalent to the operation `x*2**y` (example 25<<2 = 100). 
2. $x$ >> $y$, shifts $x$ by $y$ bits to the right; equivalent to the operation `x//2**y` (example 25>>2 = 6).
3. $x$ & $y$,  binary AND (compares bit by bit).
4. $x$ | $y$,  binary OR (compares bit by bit).
5. $x$ ^ $y$,  binary XOR (compares bit by bit).
6. \~$x$,     binary NOT (compares bit by bit), equivalent to the operation `~x = -x-1`.  

The operators <<, >>, are a fast way to multiply/divide by powers of two. For XOR (exclusive OR), it only gives one when the two bits differ; if they are equal it gives zero. The NOT (~) operation is known as the complement operation and inverts every bit, including the sign bit. As an example, consider the following operations according to the truth tables applied bit by bit, between the upper and lower bits: 
```c
       0011        1011        1010        
    OR 1000    AND 1110    XOR 1001    NOT 1000 
     = 1011      = 1010      = 0011      = 0111
```     
Just as the += operator can be used for assigning sums, the same can be done with the bitwise operators; for example, the operation `x>>=1` is equivalent to doing `x = x>>1`, more concretely, 
```python
x=1 then x>>=1 gives x=0
```
Manipulating bits can lead to very computationally efficient operations; here are a few examples:    


````python
#---- logical operations with bitwise -------------
A = int('00100011',2) # Out: 35
B = int('00101101',2) # Out: 45

print('AND', bin(A & B)) # Out: 0b100001
print('OR' , bin(A | B)) # Out: 0b101111
print('XOR', bin(A ^ B)) # Out: 0b1110
print('NOT', bin(~A),~A) # Out: -0b100100 = -36
````

````python
print('-------- shift bits right ----------')
print('A:', bin(A))
print(bin(A>>1)) # Out: '0b10001' = 35//2
print(bin(A>>2)) # Out: '0b1000'  = 35//4
print(bin(A>>3)) # Out: '0b100'   = 35//8
print('-------- shift bits left --------')
print(bin(A<<1)) # Out: '0b1000110'  = 35*2 
print(bin(A<<2)) # Out: '0b10001100' = 35*4
print(bin(A<<3)) # Out: '0b100011000'= 35*8
````

````python
# min and max of two integer values using bitwise
# Careful: this redefines (overwrites) python's built-in min() and max()
# functions for the rest of the session; if you don't want to lose the
# built-ins, use another name, e.g. bit_min, bit_max.
min = lambda x,y: y^((x^y)&-(x<y))
max = lambda x,y: x^((x^y)&-(x<y))

min(45,4), max(45,4)
````

````python
# efficiently swap the values of 2 variables
a = 1; b = 2
def swap(a,b):
    a = a^b
    b = a^b
    a = a^b
    return a,b
swap(a, b)    # note that now a = 2 and b = 1
````

````python
# Compute the Greatest Common Divisor of two numbers,
# i.e. the largest number that divides both.
def GCD(a,b):
    while True:
        a%=b; b^=a; a^=b; b^=a #b^=a^=b^=a%=b
        if (b==0): break
    return a

GCD(8,20)
````

### Masking using bitwise operators
Bitwise operators can be used to select elements of an array in python; for example a mask is defined as the condition `mask = (condition1 & condition2)`, and then it is inserted as the index of the array we want to select, i.e. `array[mask]` will only print the elements for which mask is true; in general mask is an array with logical elements, example: 


````python
# ----------mask using bitwise --------------------
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0,101,1)
y = 300 - 0.1*np.arange(0,101,1)
mask = (x >= 50) & (x <= 100) # only select elements between 50 and 100, 
# mask = (x < 40) | (x > 50)  # the parentheses matter, removing them causes an error.

print (mask) # prints an array with True or False as elements.

fig, ax = plt.subplots()
ax.scatter(x[mask], y[mask]) # note the plot starts at 50 and not at zero.
````

````python
# Convert 32-bit IEEE 754 binary to base 10 
import struct
def bin32_to_float(b):
    f = int(b,2)
    x = struct.unpack('f',struct.pack("I",f))[0]
    return x

bin32_to_float('11000100101001000010101000000000')
````

````python
# Convert 64-bit IEEE 754 binary to base-10 float 
import struct
def bin64_to_float(b):
    f = int(b,2)
    x = struct.unpack('d',struct.pack("<Q",f))[0] # "L"
    return x

# example with the number pi
bin64_to_float('0100000000001001001000011111101101010100010001000010110100011000')
````

````python
# Convert a number to its 32-bit IEEE 754 representation: 
import struct
def to_bin32(x):
    return ''.join('{:0>8b}'.format(c) for c in struct.pack('!f', x))

# Convert a number to its 64-bit IEEE 754 representation: 
def to_bin64(x):
    return ''.join('{:0>8b}'.format(c) for c in struct.pack('!d', x))

print("32-bit IEEE 754: -1313.3125 =", to_bin32(-1313.3125))
print("64-bit IEEE 754: -1313.3125 =", to_bin64(-1313.3125))
````

````python
# Example
# Alternative: Convert a float (not integers) to its 64-bit IEEE representation: 
def float_to_bin64(x):
    if x == 0: return "0"*64
    w, sign = (float.hex(x), 0) if x > 0 else (float.hex(x)[1:], 1)
    mantissa = int(w[4:17], 16) 
    exp      = int(w[18:])
    return "{}{:011b}{:052b}".format(sign, exp + 1023, mantissa)

# example with the number pi
print(to_bin64(-3.141592653589793))      # print pi, previous function
print(float_to_bin64(-3.141592653589793))# print pi, alternative function
print(float_to_bin64(-1.)) # add the point to turn the integer into a float!
````

    nbits       = 0   # count bits equal to one

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def ieee(bits, cuts, colors, texts, height=1, name="Fig"):
    fig, ax = plt.subplots(figsize=(16, .8))

    for i, b in enumerate(bits):
        ax.add_patch(Rectangle((i, 0), 1, height, fc=colors[i], ec="white"))
        ax.text(i+.5, height/2, b, ha="center", va="center", fontsize=14)

    y = height + .25
    for j, (start, end) in enumerate(zip(cuts[:-1], cuts[1:])):
        ax.plot([start, end], [y, y], color=colors[start], lw=2)
        ax.text((start+end)/2, y+.12, texts[j], ha="center", fontsize=13)

    ax.set(xlim=(-.5, len(bits)+.5), ylim=(-.1, height+.7))
    ax.axis("off")
    plt.savefig("../figures/" + name + ".png", transparent=True, format="png")


# Float32: −1313.3125
ieee("11000100101001000010101000000000",
     [0, 1, 9, 32],
     ["paleturquoise"] + ["lightgreen"]*8 + ["lightpink"]*23,
     ["Sign", "Exponent (8 bits)", "Mantissa (23 bits)"],1, "Float_example_2")

# Double: 27.56640625
ieee("0100001101000001110000110111100100110111111000001000000000000000",
     [0, 1, 12, 64],
     ["paleturquoise"] + ["lightgreen"]*11 + ["lightpink"]*52,
     ["Sign", "Exponent (11 bits)", "Mantissa (52 bits)"],1, "Double_example_2")
````
