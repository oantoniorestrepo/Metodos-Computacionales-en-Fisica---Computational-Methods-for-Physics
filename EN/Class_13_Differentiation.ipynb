# Derivatives
In some cases the function is available but the derivative is cumbersome to work out, so symbolic-computation tools can be used to obtain the derivative; in other cases only a table of data is available, with values of the function and the points at which it is evaluated, in which case the numerical derivative must be implemented. Nevertheless, there will be other cases, in physics problems, where the numerical derivative is more computationally efficient to compute than the analytical derivative. This document studies the derivative, and finally studies how the value of $h$ influences precision. Here are the topics,

## How to use sympy

Sympy is python's library for symbolic computation; it also has the functions `sin(x)`, `cos(x)`, `log(x)`, `exp(x)`, `pi`, etc. defined, but symbolically. They can be imported using,
```python 
    from sympy import *
```
or also,
```python
    import sympy as sp
```
Here we use the first form; to avoid confusion with numpy's numerical functions, these will be imported as,
```python
    import numpy as np
```
so, in numpy the functions will be: `np.sin(x)`, `np.log(x)`, `np.exp(x)`, etc. 
  
IMPORTANT: do not use `%pylab` after importing sympy as `"from sympy import *"`, since this command uses `"from numpy import *"` and numpy overrides sympy's functions; the correct way is,
```python
    %pylab
    from sympy import *
```
so, for example, `exp(x)`, `pi`, etc. are sympy (symbolic computation) and `np.exp(x)`, `np.pi` are numpy (computations with numpy arrays).

If you use, for example, `np.log(x)` symbolically or `log(x)` numerically, it will error.

For more, see [sympy tutorial](https://www.southampton.ac.uk/~fangohr/teaching/python/book/html/12-symbolic-computation.html).

## Analytical derivatives using python and its sympy library

To compute a derivative the `sympy.diff()` command is used, in either of the following two ways,
```python
    f.diff(x,n) # nth-order derivative
    diff(f,x,n)
```    
but first the variable $x$ must be defined using the `sp.Symbol()` or `sp.symbols()` command. Example: consider the function,
$$f(t)=\frac{10 \log{\left (t^{2} + t + 1 \right )}}{10 t^{3} - 20 t^{2} + t - 2}$$


````python
# NOTE: the sympy.init_printing() command prints results in latex,
#       i.e. it enables latex output for jupyter/ipython

from sympy import *
import sympy as sp
import numpy as np 
sp.init_printing(use_latex='mathjax') # nice print!!!
import matplotlib.pyplot as plt
````

````python
# Example 1)
t         = Symbol ('t')       # define a single symbol,
# x,y,z,t = symbols('x,y,z,t') # define several symbols, 
# x,y,z,t = symbols('x y z t') # also works. 
# x,y,z,t = symbols('xyzt')    # does NOT work in python 3.x (but does in 2.7)

y = 10*log(t**2+t+1)/(10*t**3-20*t**2+t-2) # create a symbolic function
# print(y), print(display(y))  # compare print() and display()
y # print the function (right-click on the function, choose Show Math As and pick a format)
````

````python
# 1) Analytical derivative of y with respect to t
dy = y.diff(t)                                
#dy = (10*log(t**2+t+1)/(10*t**3 - 20*t**2 +t-2).diff(t) # another alternative

# 2) Note another way to obtain the analytical derivative:
diff(y,t)
#diff(10*log(t**2+t+1)/(10*t**3 - 20*t**2 +t-2),t)
# the analytical derivative is more complicated to compute:
````

````python
# lambdify is used to create a numerical form of the functions:
f  = lambdify(t, y , 'numpy') # create function f(t) that can be used with numpy
df = lambdify(t, dy, 'numpy') # create a numerical function for the derivative

x = np.linspace(-1,1,1000)

plt.plot(x, f(x),label="f(x)")
plt.plot(x,df(x),label="f'(x)")

plt.legend()
plt.show()
````

## Numerical derivatives by finite differences
The numerical derivative is obtained from the definition of the limit, or also via a Taylor series, thus,

$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(\xi(x)),$$

solving for the derivative,

$$\boxed{f'(x) = \frac{f(x+h) - f(x)}{h} - \frac{h}{2}f''(\xi(x)),} $$

where the last term is the error and is of order $O(h)$. A better approximation is the derivative with error up to order $O(h^2)$, obtained from the Taylor expansion,

$$f(x+h) = f(x) + hf'(x) + \frac{h^2}{2}f''(x)  + \frac{h^3}{6}f'''(x)+ \frac{h^4}{24}f^{(4)}(\xi(x)),$$

$$f(x-h) = f(x) - hf'(x) + \frac{h^2}{2}f''(x)  - \frac{h^3}{6}f'''(x)+ \frac{h^4}{24}f^{(4)}(\xi(x)),$$

if these expressions are subtracted,

$$f(x+h)-f(x-h) = 2hf'(x) + 2\frac{h^3}{6}f'''(\xi(x)),$$ 

solving for the derivative and keeping terms up to order three,

$$\boxed{f'(x) = \frac{f(x+h)-f(x-h)}{2h}  - \frac{h^2}{6}f'''(\xi(x)).}$$ 

This last expression is known as the central derivative. Note that if instead of subtracting we add and keep terms up to order four,

$$f(x+h) + f(x-h)= 2f(x) + h^2f''(x)  + \frac{h^4}{12}f^{(4)}(\xi(x)),$$

from this expression we obtain the formula for the second derivative

$$h^2f''(x) = f(x+h) + f(x-h) -2f(x) - \frac{h^4}{12}f^{(4)}(\xi(x)),$$

finally,

$$\boxed{f''(x) = \frac{f(x+h) + f(x-h) - 2f(x)}{h^2} - \frac{h^2}{12}f^{(4)}(\xi(x)).}$$


````python
# Derivative using forward differences, from the definition of the limit
Df = lambda f, x, h: (f(x+h) - f(x))/h

# Central derivative
# obtained from the expansion of f(x+h) and f(x-h)
Dfc = lambda f, x, h: (f(x+h) - f(x-h))/(2*h)

# Numerical second derivative
D2f = lambda f, x, h: (f(x+h) - 2*f(x) + f(x-h))/h**2
````

````python
dfn = Df(f, x, 1.e-5)
plt.plot(x,dfn,label='numeric diff')
plt.plot(x,Dfc(f, x, 1.e-5),label='numeric central diff')
plt.legend()
````

## Error in the derivative and instability
Let $f'(x) = f'_{num}(x) + e(x)$, where $f'_{num}$ is an approximation of the derivative; then the absolute error for the first-order derivative is,

$$
\begin{eqnarray}
e(x)&=&|f'(x) - f'_{num}(x)|\\\\
    &=&\left|f'(x) - \frac{f(x+h) - f(x)}{h}\right|,
\end{eqnarray}
$$

and for the central (or second-order) derivative,

$$e_c(x)=\left|f'(x) - \frac{f(x+h) - f(x-h)}{2h}\right|.$$

Comparing the error for the derivative computed to first order and with the central difference, note that indeed the errors for the same value of $h$ are of order $O(h)$ and $O(h^2)$; let's look at the error in the derivative of the sine function:


````python
#Use the zoom and compare the values; to do this uncomment the following line
%matplotlib qt
x = np.linspace(0,2*np.pi,1000)
f = lambda x: 4.*np.sin(3*x + 2)

# Comparison of the central-difference and forward-difference error in f'(x)
plt.plot(x, np.abs(Df(f, x, 1.e-5) - 12*np.cos(3*x+2)),label='$O(h)$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-5) - 12*np.cos(3*x+2)),label='$O(h^2)$')
plt.legend()
````

**Task**: in the plot above, a) zoom in to compare the error value of the central difference against the first-order difference, b) change $h=10^9$ for first order and $h=10^3$ for the central difference — what do you conclude?


The smallest $h$ isn't always the best, since a balance must be struck between $h$ and the difference in the numerator (if $h$ is too small, the numerator will have a very large error, due to subtractive cancellation between two very similar numbers, $f(x+h)$ and $f(x-h)$). Let's look at how the error varies with the value of $h$:


````python
# Use the zoom and compare the values; to do this uncomment the following line
%matplotlib inline
x = np.linspace(0,2*np.pi,1000)
f = lambda x: np.sin(x)

# Using Df(x), what is the best h? what is the error?
plt.figure(2) # use another figure
plt.plot(x, np.abs(Df(f,x, 1.e-3 ) - np.cos(x)), label='$h=10^{-3 }$')
plt.plot(x, np.abs(Df(f,x, 1.e-5 ) - np.cos(x)), label='$h=10^{-5 }$')
plt.plot(x, np.abs(Df(f,x, 1.e-6 ) - np.cos(x)), label='$h=10^{-6 }$')
plt.plot(x, np.abs(Df(f,x, 1.e-9 ) - np.cos(x)), label='$h=10^{-9 }$')
plt.plot(x, np.abs(Df(f,x, 1.e-12) - np.cos(x)), label='$h=10^{-12}$')
plt.plot(x, np.abs(Df(f,x, 1.e-15) - np.cos(x)), label='$h=10^{-15}$')
plt.title('Diff error, order $O(h)$')
plt.legend()
plt.show()

# Using Dfc(x), what is the best h? what is the error?
# Solution: 
plt.figure(3) # use another figure
plt.plot(x, np.abs(Dfc(f,x, 1.e-3 ) - np.cos(x)), label='$h=10^{-3 }$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-5 ) - np.cos(x)), label='$h=10^{-5 }$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-6 ) - np.cos(x)), label='$h=10^{-6 }$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-9 ) - np.cos(x)), label='$h=10^{-9 }$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-12) - np.cos(x)), label='$h=10^{-12}$')
plt.plot(x, np.abs(Dfc(f,x, 1.e-15) - np.cos(x)), label='$h=10^{-15}$')
 
plt.title('Central diff error, order $O(h^2)$')
plt.legend()
plt.show()
````

From the two plots above, does $h = 10^{-15}$ make sense? No, the maximum error is of order $\sim 0.15$ (or $15$%) compared to the analytical derivative.
 
 What is the best $h$?
 
+   for `Df(x)`  the best $h$ is $10^{-9}$ with maximum error $e \sim 10^{-7}$.
+   for `Dfc(x)` the best $h$ is $10^{-5}$ with maximum error $e \sim 10^{-11}$.
