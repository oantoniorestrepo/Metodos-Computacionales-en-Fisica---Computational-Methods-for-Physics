## Random number generation and descriptive statistics 
### Probability theory concepts

**Probability** is a concept used daily; in general terms it can be seen as a quantification of our expectation of obtaining a result from an event in an experiment; more quantitatively, if in an experiment of $n$ trials $f_i$ observations of an event $i$ occur, then the probability $p_i$ of observing event $i$ is,
<!---
más cuantitativamente, suponga que hay $m$ eventos distintos a observar en un experimento y que este se repite $n$ veces, entonces si en el experimento ocurren $f_i$ observaciones de un evento $i$, la probabilidad $p_i$ de observar el evento $i$ es,
--->
$$p_i=\frac{f_i}{n}$$

note that, $\sum_{i=1}^m f_i = n,$ so $\sum_{i=1}^m p_i=1,$ where $m$ is the total number of distinct events to observe, different from $n$, the number of repetitions in the experiment; also, since $f_i\leq n$, then $0\leq p_i\leq 1$. The mean value (`np.mean()`) is defined by,

$$
\mu   \equiv \langle x\rangle=\frac{1}{n}\sum_{i=1}^n x_i=\sum_{i=1}^m x_i p_i\\\\
$$

where for the second sum we note that each $x_i$ repeats $f_i$ times. To find the error, we look at how much each quantity $x_i$ differs from the mean value $\mu$, then compute the average of these squared differences; this is known as the standard deviation (`np.std()`), i.e.,

$$
\begin{align}
\sigma&\equiv\sqrt{\langle (x-\mu)^2\rangle}=\sqrt{\frac{1}{n}\sum_{i=1}^n (x_i-\mu)^2}=\sqrt{\sum_{i=1}^m (x_i-\mu)^2p_i},\\
\sigma&=\sqrt{\langle x^2\rangle-\langle x\rangle^2},
\end{align}
$$

where $\langle x^2\rangle=\sum_{i=1}^n x_i^2$. Also $\sigma^2=\,$var$(x)$, known as the variance (`np.var()`), which equals the mean value $\langle (x-\mu)^2\rangle$, i.e. the variance measures how much each quantity $x_i$ deviates from the mean $\mu$. If the variable $i$ takes continuous values, $f_i\rightarrow f(x)$, then the probability of obtaining values (realisations) of $x$ in the interval $[x,x+dx]$ is, 

$$p=f(x)dx\approx f(x_i)\Delta x$$ 

where $f(x)$ is known as the probability density; the normalisation condition is also imposed,

$$\int_{-\infty}^\infty f(x)dx=1,$$

so again $p$ takes values between zero and one.

**Cumulative function** or **distribution function**, gives us the probability of finding a value equal to or smaller than $x$, where $F(x)$ takes uniformly distributed values on the interval $[0,1]$; it is defined by the integral,

$$F(x) = \int_{-\infty}^xf(x')dx',\quad \text{equivalent to}\quad F(k)=\sum_{i=1}^{k} p_i,\quad \text{with}\quad k\leq m.$$

Note that if $x=\infty$ then $F(\infty)=1$; in the discrete case the integral is replaced by a sum.


**Binomial distribution**, a basic element of descriptive statistics, given by,

$$f(x;n,p) = \Pr(X = x) = \binom n x  p^x(1-p)^{n-x} = \frac{n!}{x!(n-x)!}p^x(1-p)^{n-x},$$

where $\Pr(X = x)$ is the probability that an independent event occurs $x$ times in $n$ trials, $p$ is the individual probability that an event occurs and $1-p$ that it does not. For example, suppose that in an experiment the probability that a certain result comes up is $p=0.25$, so the probability that it doesn't is $1-p=0.75$; if repeated $10000$ times, $\Pr(X = 100)$ gives us the probability of obtaining that result 100 times.

**Gaussian distribution**, this distribution is obtained as a limiting case of the binomial distribution as $n\to\infty$ (although even for values as low as $n=10$ it represents the properties of the binomial distribution well); it is represented by the function,

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2} } \; e^{ -\frac{(x-\mu)^2}{2\sigma^2} }.$$

Where $\sigma=\sqrt{np(1-p)}$ is the standard deviation and $\mu=np$ the mean value.  

**The Poisson distribution**, another special case of the binomial distribution, occurs when $n\to\infty$ but $p\to 0$ such that $np$ is constant,

$$f(x) = e^{-\mu}\frac{\mu^x}{x!}.$$


## Generating random numbers
Before showing how to generate random numbers with a given distribution, let's first look at how uniformly distributed random numbers are generated by the computer.

### Linear congruential method
This is the most common method for generating random numbers on an interval $[0,M-1]$; a random number is defined as $r_i$ and computed using

$$r_i=(ar_{i-1}+c)\hbox{mod}(M),$$

where mod($M$) is the modulo of $M$, $r_1$ is known as the seed and the other $r_i$ are generated from this number; if we want numbers on the interval $[0,1)$ we divide by $M$ (note the interval is open at one, since that value is never generated). It's recommended to use python's function to generate random numbers

    numpy.random.rand(n)
    numpy.random.random(n)
    
these generate a vector of dimension $n$ with random numbers $r$ on the interval $[0,1)$. Number generation on the computer is "*pseudo*" random, since due to its architecture it cannot create truly random numbers.

In general, every pseudo-random number generator is limited to generating uniform distributions on the interval $[0,1)$; so how do you generate other distributions from this one? The following sections explain how:


### Uniform distribution
To generate a uniform distribution on $[a,b)$ a linear mapping is done,

$$x = a+(b-a)r$$

which follows the distribution,

$$f(x)=\frac{1}{b-a}, \quad\text{for}\quad  a<x<b$$

Note that the previous mapping of $x$ can also be computed using the cumulative function on $[a,x]$, noting that $F(x)=r$ is a random number between zero and one; then solving for $x$ as a function of $r$ recovers the previous expression for $x$.

### Exponential distribution
Neutral particles like photons travel in a straight line between points of interaction with matter; the mean distance between these points is known as the *mean free path*, and it follows an exponential distribution defined by

$$f(x) = \frac{1}{\lambda}e^{-\frac{x}{\lambda}}$$ 

In this case, $\lambda=1/\sigma$, is the inverse of the cross section (cm$^{-1}$), which is a measure of the interaction between particles fired at a scattering centre (interaction point). The idea is to simulate the generation of these particles according to this exponential distribution.
In general, to generate random numbers with an exponential distribution
such that $0<x<\infty$ we use the formula

$$x = -\lambda\ln(1 -r_i).$$

This formula is obtained from the distribution function 

$$F(x)=r_i=\int_0^x \frac{1}{\lambda}e^{-\frac{x}{\lambda}} dx.$$

since $F(x)$ guarantees uniformly distributed numbers on the interval $[0,1]$.

### Gaussian distribution
There are several ways to generate a Gaussian distribution: with the Box-Muller method, by the central limit theorem, and by the Von Neumann method:

**1)** [**The Box-Muller method**](https://en.wikipedia.org/wiki/Box%E2%80%93Muller_transform):
let $r_1$ and $r_2$ be independent random variables uniformly distributed on the interval $[0, 1)$; then

$$x =\sqrt{-2 \ln r_1} \cos(2 \pi r_2)$$

has a Gaussian distribution. In python there are two functions that generate numbers with Gaussian distribution from the Box-Muller formula,
```python
x = numpy.random.randn(n)  # distribution with standard deviation (sigma) one and centred at zero.
x = numpy.random.normal(size=n) # similar
```
since $x$ has a standard deviation of one and mean value zero, if you want to generate a number $x$ with standard deviation 
$\sigma$ and mean value $\mu$, the scaling is done,

$$r=\mu+x\sigma.$$

The opposite operation, $x=\frac{r-\mu}{\sigma}$, is known as *standardisation*.
Note: don't confuse `randn()` with `rand()`, which generates a uniform distribution.

**Example**:


````python
import matplotlib.pyplot as plt
from numpy.random import normal

n =10000
# 1) Gaussian using python's libraries 
# subplot(row,column,position), e.g.: "234" is a 2x3 grid, 4th subplot.
plt.subplot(121) #(211)
plt.title("Gaussian")
plt.xlabel("x")
plt.ylabel("Distribution")

x = normal(size=n)
plt.hist(x, bins=100,density=True)


# 2) Gaussian using the Box-Muller method
plt.subplot(122) #(211)
plt.title("Gaussian")
plt.xlabel("x")
#plt.ylabel("Distribution")

r1= random.random(n)
r2= random.random(n)
x = sqrt(-2*log(r1))*cos(2*pi*r2) # x is an array

plt.hist(x, bins=100,density=True, color='r')
show()
````

**2) Central limit theorem**:
This theorem gives a simple (though less efficient) way to generate numbers with a Gaussian distribution. The theorem states that if a sequence of numbers ${r_i}$ generated independently of each other, then their sum 

$$x = \sum_{i=1}^Nr_i$$

follows a Gaussian distribution

$$f(x) = \frac{1}{\sqrt{2\pi\sigma^2} } \; e^{ -\frac{(x-\mu)^2}{2\sigma^2} },\,\, \mu=N\langle r\rangle ,\,\, \sigma^2=N(\langle r^2\rangle -\langle r\rangle ^2).$$

To obtain the distribution, this calculation is repeated $M$ times, giving M values $x_j$ where each $x_j$ is the sum of $N$ values $r_i$.


````python
# Gaussian using the central limit theorem
M = 10000
N = 1000

#-- gaussian, central limit theorem --
h=zeros(M)
for i in range(M):
    # generate N events with different distributions:  
    # r=rand(N)+randn(N)+exponential(size=N)+poisson(1,size=N)

    r=random.rand(N) # generate N events with uniform distribution  
    x=sum(r)         # sum of independent events
    h[i]=x           # store x in the histogram h
    
h=( h - h.mean() )/h.std() # standardisation
hist(h,bins=100, density=True)    

#-- theoretical gaussian distribution function -
f = lambda t: 1./sqrt(2*pi)*exp(-t**2./2.) 

t = linspace(-4,4,N)
plot(t,f(t))
show()
````

Although this method generates random numbers with a Gaussian distribution, it isn't efficient, since rejection loses many numbers and computation time.

In experimentation, when a quantity is measured there's an error associated with it; the error is given by the standard deviation, so, let $y(x)$ be the quantity to compute; then from the experiment (repeated $n$ times for a fixed value of $x$) we find the value of $y$ should be, 

$$y=\mu\pm\sigma,$$ 

if the errors are independent of each other, then by the central limit theorem the total error will be Gaussian; this must be done for each value of $x$. For example if $T=y(x)$ represents the temperature at a point $x$ on a bar between two heat reservoirs (one end at $0^\circ$ C and the other at $100^\circ$ C), if the temperature is measured at the midpoint $n$ times, it will differ at each measurement, so we can say that $T(x)=\mu(x)\pm\sigma(x)$; the error will be the sum of many independent errors and therefore Gaussian (electrical fluctuations, non-uniform resistances, poor fridge insulation, so the reservoirs don't maintain a constant temperature, environmental factors, air currents, an uncalibrated thermometer, etc).

### Poisson distribution
To generate integers $k$ with a Poisson distribution, 
$$f(k) = e^{-\mu}\frac{\mu^k}{k!},$$
where $k = 0, 1, 2, ...$ times in the interval, we do the following steps:

 1. Initialise: $k = 0, A = 1$.
 2. Generate a uniform random number $u$ on $[0,1]$, define $A = uA$.<br> 
    If $A \leq \exp(−\mu)$ accept $k$ and stop, otherwise:
 3. Increment $k = k + 1$ and go back to step 2.
 
This algorithm gives us the probability of observing $k$ events in a given interval:


````python
#   *** Poisson distribution  *** 

n  = 10000 # total number of events
mu = 2     # rate of events (average number of events per interval)

h=[]
for i in range(n):
   k=0; A=1 
   while 1:  
      u = random.random()  # generate random num in [0,1]
      A = u*A            
    
      if (A <= np.exp(-mu)):
        h = np.append(h,k) # accept k
        break              # stop
      else:
        k = k + 1          # increment k
print(h.min(),h.max()) 

plt.hist(h,bins=50,weights=np.ones(n)/n)#, normed=True) 
show()
````

=&\sum_{i=1}^N \Delta\mathbf{r}_i.
\end{align}
$$
si se toma el producto punto $(\mathbf{r}-\mathbf{r}_0)^2$, y se considera que el movimiento es en 2D con $\mathbf{r}=x\mathbf{\hat i}+y\mathbf{\hat j}$, e igualmente para cada $\Delta\mathbf{r}_i$
--->
The 2D random walk is generated from consecutive steps on the interval $[-1,1]$ (see the Landau-Páez book for the theory).

Code that simulates the random motion of a particle in 2D:


````python
N = 100         # steps
x=0;  y=0       # current step position
x0=0; y0=0      # previous step position
plot(x0,y0,"ro")# origin

for i in range(N):
   x += ( random.rand() - 0.5 )*2. # -1 =< x =< 1
   y += ( random.rand() - 0.5 )*2. # -1 =< y =< 1
   
   plot([x0, x],[y0, y],'k-')
   plot(x,y,"b>")
   #plt.arrow(x0, y0, x-x0, y-y0,head_width=0.2, head_length=0.3, fc='k', ec='b')
   x0=x; y0=y
plt.title("2D random walk")
show()
````

````python
N = 100         # steps
x=0;  y=0       # current step position
x0=0; y0=0      # previous step position
plot(x0,y0,"ro")# origin
scalar = 1
for i in range(N):
   dx = ( random.rand() - 0.5 )*2. # -1 =< x =< 1
   dy = ( random.rand() - 0.5 )*2. # -1 =< y =< 1
   # normalise
   norm = np.sqrt(dx**2 + dy**2)

   x += scalar*dx/norm 
   y += scalar*dy/norm
    
   plot([x0, x],[y0, y],'k-')
   plot(x,y,"k>")
   #plt.arrow(x0, y0, x-x0, y-y0,head_width=0.2, head_length=0.3, fc='k', ec='b')
   x0=x; y0=y
plt.title("2D random walk")
show()
````

### Exponential decay

Python code for simulating exponential decay (see the Landau-Páez book for the theory):        


````python
# Decay.py spontaneous decay simulation
import matplotlib.pyplot as plt
import numpy as np

lamda = 1.e-1 # Decay constant
N = 1000000
t = 0
tt = []
Nt = []
DeltaNt = []

while N > 1:    
    DeltaN = 0
    for i in range(N):
        if ( np.random.rand() < lamda ): DeltaN += 1

    t += 1.
    N -= DeltaN
    
    tt= np.append(tt,t)
    Nt= np.append(Nt,N)
    DeltaNt = np.append(DeltaNt,DeltaN)
    
plt.subplot(121) #(211)
plt.semilogy(tt,Nt)
plt.title("Exponential decay N(t)")
plt.xlabel("t")

plt.subplot(122) #(212)
plt.semilogy(tt,DeltaNt/Nt) # probability P = N/dN/dt = -lambda
plt.title(r"$P(t) = \Delta N(t)/N(t)/\Delta t = -\lambda$")
show()
````

### Poisson processes

Consider the following physical situations:
1. We want to compute the efficiency of the number of photons reaching a detector (for example a telescope's) in each time interval.
2. Suppose that in a thousand years, 100 supernovae are observed in a region of the universe, and we want to know the time between two supernovae and with what probability.
3. It's known that about 13000 earthquakes occur on Earth per year, i.e. one every ~40 min; if one occurs now, when will the next one occur?
4. We want to compute the diffusion times of a particle in a medium.

All these problems refer to situations where you can keep a count of a certain number of events that occur at a given frequency in time but are random in nature (without any structure); this class of problems is known as [Poisson processes](http://preshing.com/20111007/how-to-generate-random-timings-for-a-poisson-process/). 

In general *Poisson processes occur when there are rare events that occur individually and randomly* (they tend to occur at a constant average rate when seen as a group, and there's no memory or correlation between events). 
For example, in an experiment 10000 data points are obtained each hour, i.e. on average $\lambda = 10000/(3600\text{ s })\approx2.777$ s$^{-1}$ data points are obtained each second, meaning each event is generated on average every $3600/10000=0.36$ s. In general the frequency can be defined as,

$$\lambda=\frac{\text{number of events}}{\text{total time}},$$ 

which gives a measure of the frequency of events, or the average rate of events per unit time. 
Now let's ask the following question: what is the probability $P$ of having an event in the next $2.0$ s if one just occurred right now? To answer the question we use the Poisson distribution function,

$$F(N)=\sum_{k=0}^Ne^{-\mu}\frac{\mu^{k}}{k!},$$ 

which gives the probability of observing a number of events equal to or smaller than $N$ for a time $t$. Since Poisson processes are processes that occur in time, the individual probability that an event occurs in a time interval $\Delta t$ is given by $p=\lambda \Delta t$, and if there are $N$ processes in a time $t$ then $\mu=Np=N\lambda \Delta t$, which gives, 

$$\mu=\lambda t.$$

From the distribution function, the probability of having no event in $\Delta t$ occurs when $N=0$, or $F(0)=e^{-\lambda \Delta t}$, and the probability that one does occur is, $P(\Delta t)=1-F(0),$ which gives, 

$$ \boxed{P(\Delta t) = 1 - e^{-\lambda \Delta t}.}$$

We infer from this formula that as time passes the probability tends to one, so the question isn't whether the event will occur or not, but when it will occur. So, 

$$P(2.0 s)= 1 - e^{-2.777\times 2}\approx0.996,$$

is the answer. If we solve for the time from this formula and recall that $F$ has uniformly distributed values on $[0,1]$, then, to generate the time at which a Poisson process will occur, 
a random number $r_i$ is generated between $[0,1)$ and the time is computed as,

$$\boxed{\Delta t = \dfrac{-\ln(1- r_i)}{\lambda}.}$$

Note you could use $r_i$ instead of $(1-r_i)$, but the latter is more convenient since it avoids passing zero to the logarithm (since the $r_i$ generator never produces 1). 
In the following python example, the first plot shows the exponential nature of the system, the second the time line between $10$ consecutive events, and it's also shown that for $N$ consecutive events, 

$$\lambda \approx\frac{1}{N}\sum_{i=1}^N\dfrac{-\ln(1- r_i)}{(1/\lambda)}.$$ 

**Example**: python program that simulates the occurrence of events with time $\Delta t_i$ as a function of the rate $\lambda$ (note that in the sum the inverse of $\lambda$ is used):


````python
# Poisson_process.py
# To get time for obtaining an event if given the rate lambda.
# Note that we cannot pass zero to np.log(), but we avoid that 
# by subtracting the result of random.rand():

steps = 100
lamda =  10000/3600 # fecuency of events

def delta_t(lamda):
    return -log(1.0 - random.rand())/lamda # rate Parameter

#------------ 1) plot showing exponential behavior -----------
plt.subplot(211) #(121)
t=arange(1,steps)
plot(t,delta_t(t),'o') 

#---------2) plot of the N consecutive delta t steps ---------
plt.subplot(212) #(122)
t=[]
for i in range(10):
    t.append(delta_t(1/lamda))

plt.plot(t, zeros(10)) 
plt.plot(t, zeros(10), 'o')

#-- tests to verify that the average time returned is lamda --
Sum_steps = sum([delta_t(1/lamda) for i in range(steps)])/steps

print ("The sum should be approx lambda:",lamda, Sum_steps)
````

### Linear regression: least squares 
Normally in experiments we want to check whether the data obtained fits a theoretical model, so a function $g(x_i,\{a_1,\dots,a_{N_p}\})$ is proposed with $N_p$ theoretical parameters $\{a_1,\dots,a_{N_p}\}$ to be fitted according to $N_D$ experimental data points $(x_i,y_i\pm\sigma_i)$ (where the $\sigma_i$ are the error of the $y_i$; if the errors aren't given, $\sigma_i=1$ is taken). 
A common method used is least squares, which, to find the curve that best fits the $N_D$ data points $(x_i,y_i\pm\sigma_i)$, computes the squared error, defined as the sum of the data obtained minus the proposed theoretical function evaluated at each point $x_i$, i.e.

$$\chi^2=\sum_{i=1}^{N_D}\left(\frac{y_i-g(x_i,\{a_1,\dots,a_{N_p}\})}{\sigma_i}\right)^2,$$

to find the parameters $\{a_i\}$ that best fit, the error is minimised

$$\frac{\partial \chi^2}{\partial a_k}=0 \iff\sum_{i=1}^{N_D}\left(\frac{y_i-g(x_i,\{a_1,\dots,a_{N_p}\})}{\sigma_i}\right)\frac{\partial g}{\partial a_k}=0$$

This gives a system of $N_p$ equations where the values of the $N_p$ parameters $a_k$ must be found. If the function is $g(x) = a_1 + a_2x $ this gives a system of 2 equations to solve; likewise if the function is $g(x) = a_1 + a_2x + a_3x^2$ the system has 3 parameters to find. In general, in the case of polynomial functions the equations are linear, but if the equations are exponential, $g(x)=be^{ax}$ or $g(x)=bx^a$), or of another form, for example $g(x)=\frac{a_1}{(x^2-a_2^2)^2+a_3^2}$, non-linear systems of equations are obtained, and only approximate values of the ${a_k}$ can be found. The fit is considered good if $\chi^2\approx N_D-N_p$.

The following numpy function can be used for least squares:

    numpy.linalg.lstsq(a, b, rcond=-1)
 
**Least-squares fit with a quadratic function**:
The following algorithm computes the least-squares fit for a parabola, 

$$g(x) = a_1 + a_2x + a_3x^2$$

Since there are 3 parameters $a_i$, this gives a system of 3 equations (see the Landau-Páez book for the theory, chap 8), 

$$ Sa_1 + S_xa_2 + S_{xx}a_3 =S_y,$$ 
$$ S_xa_1 + S_{xx}a_2 + S_{xxx}a_3 =S_{xy},$$
$$ S_{xx}a_1 + S_{xxx}a_2 + S_{xxxx}a_3 =S_{xxy};$$

then the linear system of equations $A{\bf a} =\bf b$ is solved. 
Note this algorithm also works if $g(x) = a_1 + a_2x $.
Python program:


````python
# Fit.py Linear square fit 
# from Landau-Paez book, see chapter 8 
from numpy import *
from numpy.linalg import *
import matplotlib.pyplot as plt


t = arange( 1.0,2.0,0.01)                       # Curve
Nd = 7                                          # Number of data
x = array([1.,1.05,1.15,1.32,1.51,1.68,1.92])   # Given x values
y = array([0.52,0.73,1.08,1.44,1.39,1.46,1.58]) # Given y values
sig=array([0.1,0.1,0.2,0.3,0.2,0.1,0.1])        # errorbar lenghts

plt.plot( x , y ,'bo') # Plot data 
plt.errorbar(x,y,sig)  # Plot error bars
plt.title('Linear least square fit') 
plt.xlabel( 'x' ) 
plt.ylabel( 'y' )
plt.grid( True )

A = zeros((3,3))
bvec = zeros((3,1))
ss = sx = sxx = sy = sxxx = sxxxx = sxy = sxy = sxxy = 0.

for i in range(Nd):
   sig2 = sig[i]*sig[i]   
   ss  += 1./sig2;          sx += x[i]/sig2;        sy += y[i]/sig2
   rhl  = x[i]*x[i];       sxx += rhl/ sig2;      sxxy += rhl*y[i]/sig2
   sxy += x[i]*y[i]/sig2; sxxx += rhl*x[i]/sig2; sxxxx += rhl*rhl/sig2

A = array([[ss,sx,sxx],[sx,sxx,sxxx],[sxx,sxxx,sxxxx]])
bvec = array([sy,sxy,sxxy])

ai = solve(A, bvec)   # Solution via elimination

print('x Matrix via direct')
print(ai)
print('Fit Parabola Final Results\n')
print('y(x) = a0 + a1 x + a2 xˆ2') 
print('a0 = ' , ai[0])
print('a1 = ' , ai[1])
print('a2 = ' , ai[2], '\n')
print(' i xi yi yfit ')

for i in range(Nd):
    s = ai[0] + ai[1]*x[i] + ai[2]*x[i]*x[i]
    print(" %d %5.3f %5.3f %8.7f \n" %(i, x[i] , y[i] ,s) )

# Red line is the fit, red dots the fit saty[i]
curve  = ai[0] + ai[1]*t + ai[2]*t**2.
points = ai[0] + ai[1]*x + ai[2]*x**2.
plt.plot(t, curve, 'r' , x , points , 'ro')

plt.show()
````

### Law of large numbers
Consider the average value of $n$ quantities

$${\overline {X}}_{n}={\frac {1}{n}}(X_{1}+\cdots +X_{n}) $$

The law of large numbers states that this value converges to the expected value $\mu$ as $n$ tends to infinity, i.e. 

\begin{matrix}{}\\
{\overline {X}}_{n}\,\to \,\mu \qquad {\textrm {as}}\qquad n\to \infty .\\{}
\end{matrix} 

For example, the expected value when rolling a die is given by

$${\frac{1+2+3+4+5+6}{6}}=3.5$$ 

If the die is rolled many times and ${\overline {X}}_{n}$ is computed, we see this value tends to $3.5$ as $n$ grows. Likewise for the probability, which tends to $1/6\approx0.16667$. Let's see this in the following numerical experiment:


````python
#------------ LAW OF LARGE NUMBERS ---------------------------
# Here the law of large numbers is applied to rolling a
# die (examples 1. and 2.).

import numpy as np
from matplotlib.pyplot import *

## --------------------1) EXAMPLE -----------------------------
# The expected value of rolling a die n times
# tends to 3.5 as n grows large:

figure(1)

n = 10000             # max number of rolls
x = np.arange(1,n+1)  # x axis
for i in range(10): # repeat 10 times
  r = np.random.randint(1,7, size=n) # random array on [1,6] 
  p = np.cumsum(r)/x # cumulative sum 
  plot(x,p)         

plot([0,n],[3.5,3.5],'k--',label='3.5%') # limit value

plt.xlabel("Number of rolls")
plt.ylabel("Expected value")
legend()

# --------------------2) EXAMPLE ------------------------------
# Computing the probability as n grows large:
figure(2)

prob = 1/6. # probability for a die: 0.16666667
for i in range(10): # repeat 10 times
  r = (np.random.rand(n) < prob).astype(np.int32)
  p = np.cumsum(r)/x
  plot(x,p)         

plot([0,n],[prob,prob],'k--',label='P=1/6') # limit value

plt.xlabel("Number of rolls")
plt.ylabel("Probability")
legend()
````

### Step-by-step histogram
To make a histogram we normally use the command `plt.hist(h, bins=m)`; here a histogram is built step by step:<br>
1) each value of $x$ is converted to an integer index: subtract the minimum of $x$ to get positive indices starting at zero, then take the integer part of dividing by $\Delta x=(x_{max}-x_{min})/m$, i.e.,

$\quad i=\text{int}\left(\dfrac {x-x_{min}}{\Delta x}\right)$

index with values $i=0,1,2...,m,$ where $m$ is the maximum number of bins (bars of height h).<br>
2) Each value $x_i$ belongs to the histogram $h_i$, so every time a value of $x$ falls in the interval $[x,x+\Delta x]$ the histogram is incremented by one,

$\quad h_i = h_i + 1$.<br>

These three steps are repeated for all $n$ values of $x$. Note that step 2) generates $m+1$ indices, where index $m$ corresponds to the element $x_{max}$, so it must be placed in bar $h_{m-1}$ since there are only $m$ bars; one trick is to simply increase $x_{max}$ by a fraction (for example, multiply by $1.0001$) when computing $\Delta x$ to force $x_m$ to fall in $h_{m-1}$.<br>
3) $x$ is discretised into fractions of $\Delta x$ as,

$\quad x_i=i\Delta x = \text{int}\left(\dfrac {x-x_{min}}{\Delta x}\right)\Delta x\quad\longrightarrow\quad$ value of $x$ at position $i$.<br>

So every value of $x$ in the interval $[x,x+\Delta x]$ has been approximated by the discrete value $x_i$.<br> 
4) Finally the normalisation is done, to obtain the probability density,

$\quad p_i=\dfrac{h_i}{n}\quad\longrightarrow\quad f(x_i)=\dfrac{h_i}{n \Delta x}=\dfrac{p_i}{\Delta x}$.


````python
# histogram implementation:
n = 10000 
bins = 100 
h = zeros(bins) 
x = randn(n) 
delta = (1.0001*x.max() - x.min())/bins

for i in range(n): 
    xi = int((x[i]-x.min())/delta)
   # if xi == bins: xi = xi-1, # xi produces m+1 bins, but it's more efficient 
    h[xi] += 1                 # to multiply xmax by 1.0001 in delta.
        
xi = arange(x.min(), x.max(),delta)     
# plot(xi,h/(n*delta))
bar(xi,h/(n*delta), width=delta) 
````

````python
# 1) *** Random walk ***
N = 10000    # number of steps per trajectory
M = 100      # average over M trajectories
R = zeros(N) 

for j in range(M):
    x = 0. 
    y = 0. 
    for i in range(N): 
        x    += (rand() - 0.5)*2.*2.**.5 
        y    += (rand() - 0.5)*2.*2.**.5 
        R[i] += sqrt(x*x + y*y) 

rootN = sqrt(arange(N))          
plot(rootN, R/M)
ylabel(r'$R_{rms}$')
xlabel(r'$\sqrt{N}$')
show()
````

````python
# 2) *** Random walk version 2 (slower) ***
Nmax = 100       # steps
M = 1000         # average over M trajectories.

R = zeros(M)
Rrms = zeros(Nmax)

for N in range(Nmax):
    for k in range(M):

        # ---- one trajectory -----------------------
        x=0;  y=0      # current step position
        for i in range(N):
            xi = (rand() - 0.5)*2.     # -1 =< x =< 1
            yi = (rand() - 0.5)*2.     # -1 =< y =< 1
            norm = (xi*xi + yi*yi)**.5 # normalisation
            x += xi/norm
            y += yi/norm
        # --------------------------------------------
        R[k] = x*x + y*y # includes correlations <xi*xj>.
        
    Rrms[N] = sqrt(R.mean())
    
N = arange(0,Nmax)
sqrtN = sqrt(N)

plot(sqrtN, Rrms)
ylabel(r'$R_{rms}$')
xlabel(r'$\sqrt{N}$')
show()
````
